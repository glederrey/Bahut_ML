### Étape 1. Importer les bibliothèques nécessaires

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
```

### Étape 2. Importer le dataset depuis cette [adresse](https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/09_Time_Series/Apple_Stock/appl_1980_2014.csv)

### Étape 3. L'assigner à une variable appelée `apple`

```python
url = 'https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/09_Time_Series/Apple_Stock/appl_1980_2014.csv'
apple = pd.read_csv(url)
apple.head()
```

### Étape 4. Vérifier le type des colonnes

```python
apple.dtypes
```

### Étape 5. Transformer la colonne `Date` en type datetime

```python
apple['Date'] = pd.to_datetime(apple['Date'])
apple['Date'].head()
```

### Étape 6. Définir la colonne `Date` comme index

```python
apple = apple.set_index('Date')
apple.head()
```

### Étape 7. Y a-t-il des dates dupliquées ?

```python
apple.index.is_unique
```

### Étape 8. Il semble que l'index commence par la date la plus récente. Trier pour que la première entrée soit la plus ancienne.

```python
apple = apple.sort_index(ascending=True)
apple.head()
```

### Étape 9. Obtenir le dernier jour ouvrable de chaque mois

```python
apple_month = apple.resample('BM').mean()
apple_month.head()
```

### Étape 10. Quelle est la différence en jours entre la première et la dernière date ?

```python
(apple.index.max() - apple.index.min()).days
```

### Étape 11. Combien de mois de données avons-nous ?

```python
len(apple_month.index)
```

### Étape 12. Tracer la valeur `Adj Close`. Définir la taille de la figure à 13,5 x 9 pouces

```python
fig = px.line(apple, x=apple.index, y='Adj Close', title='Valeur ajustée des actions Apple')
fig.update_layout(width=1350, height=900, xaxis_title='Date', yaxis_title='Adj Close')
fig.show()
```

### Étape 1. Importer les bibliothèques nécessaires

```python
import pandas as pd
import numpy as np
```

### Étape 2. Importer le dataset depuis cette [adresse](https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data)

### Étape 3. L'assigner à une variable appelée `wine`

```python
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data'
wine = pd.read_csv(url)
wine.head()
```

### Étape 4. Supprimer les colonnes 1, 4, 7, 9, 11, 13 et 14

```python
wine = wine.drop(wine.columns[[0, 3, 6, 8, 11, 12, 13]], axis=1)
wine.head()
```

### Étape 5. Attribuer les noms de colonnes suivants :

Les attributs sont (donnés par Riccardo Leardi, riclea '@' anchem.unige.it) :  
1) alcohol  
2) malic_acid  
3) alcalinity_of_ash  
4) magnesium  
5) flavanoids  
6) proanthocyanins  
7) hue

```python
wine.columns = ['alcohol', 'malic_acid', 'alcalinity_of_ash', 'magnesium', 'flavanoids', 'proanthocyanins', 'hue']
wine.head()
```

### Étape 6. Définir les valeurs des 3 premières lignes de `alcohol` comme NaN

```python
wine.iloc[0:3, 0] = np.nan
wine.head()
```

### Étape 7. Définir maintenant les valeurs des lignes 3 et 4 de `magnesium` comme NaN

```python
wine.iloc[2:4, 3] = np.nan
wine.head()
```

### Étape 8. Remplir les valeurs NaN avec le nombre 10 pour `alcohol` et 100 pour `magnesium`

```python
wine['alcohol'].fillna(10, inplace=True)
wine['magnesium'].fillna(100, inplace=True)
wine.head()
```

### Étape 9. Compter le nombre de valeurs manquantes

```python
wine.isnull().sum()
```

### Étape 10. Créer un tableau de 10 nombres aléatoires jusqu'à 10

```python
random = np.random.randint(10, size=10)
random
```

### Étape 11. Utiliser les nombres aléatoires générés comme index et attribuer la valeur NaN à chaque cellule correspondante

```python
wine.loc[random, 'alcohol'] = np.nan
wine.head(10)
```

### Étape 12. Combien de valeurs manquantes avons-nous ?

```python
wine.isnull().sum()
```

### Étape 13. Supprimer les lignes contenant des valeurs manquantes

```python
wine = wine.dropna(axis=0, how='any')
wine.head()
```

### Étape 14. Afficher uniquement les valeurs non nulles de `alcohol`

```python
wine['alcohol'].dropna()
```

### Étape 15. Réinitialiser l'index pour qu'il commence à 0

```python
wine = wine.reset_index(drop=True)
wine.head()
```

```python
csv_url = 'https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/04_Apply/Students_Alcohol_Consumption/student-mat.csv'
df = pd.read_csv(csv_url)
df.head()
```

### Étape 4. Pour les besoins de cet exercice, découpez le DataFrame de 'school' jusqu'à la colonne 'guardian'.

```python
stud_alcoh = df.loc[: , "school":"guardian"]
stud_alcoh.head()
```

### Étape 5. Créez une fonction lambda qui mettra les chaînes de caractères en majuscule.

```python
capitalizer = lambda x: x.capitalize()
```

### Étape 6. Mettez en majuscule les colonnes Mjob et Fjob.

```python
stud_alcoh['Mjob'].apply(capitalizer)
stud_alcoh['Fjob'].apply(capitalizer)
```

### Étape 7. Affichez les derniers éléments du jeu de données.

```python
stud_alcoh.tail()
```

### Étape 8. Avez-vous remarqué que le DataFrame original est toujours en minuscules ? Pourquoi ? Corrigez cela et mettez en majuscule Mjob et Fjob.

```python
stud_alcoh['Mjob'] = stud_alcoh['Mjob'].apply(capitalizer)
stud_alcoh['Fjob'] = stud_alcoh['Fjob'].apply(capitalizer)
stud_alcoh.tail()
```

### Étape 9. Créez une fonction appelée majority qui retourne une valeur booléenne dans une nouvelle colonne appelée legal_drinker (Considérez qu'un majeur a plus de 17 ans).

```python
def majority(x):
    if x > 17:
        return True
    else:
        return False
```

```python
stud_alcoh['legal_drinker'] = stud_alcoh['age'].apply(majority)
stud_alcoh.head()
```

### Étape 10. Multipliez chaque nombre du jeu de données par 10.

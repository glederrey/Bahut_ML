### Étape 1. Importer les bibliothèques nécessaires

```python
import pandas as pd
import plotly.express as px
from collections import Counter
```

### Étape 2. Importer le dataset depuis cette [adresse](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv).

### Étape 3. L'assigner à une variable appelée `chipo`.

```python
url = 'https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv'
chipo = pd.read_csv(url, sep='\t')
```

### Étape 4. Afficher les 10 premières entrées

```python
chipo.head(10)
```

### Étape 5. Créer un histogramme des 5 articles les plus achetés

```python
# Obtenir la série des noms des articles
x = chipo.item_name

# Utiliser la classe Counter pour créer un dictionnaire avec les clés (articles) et leur fréquence
letter_counts = Counter(x)

# Convertir le dictionnaire en DataFrame
df = pd.DataFrame.from_dict(letter_counts, orient='index', columns=['count'])

# Trier les valeurs par ordre décroissant et sélectionner les 5 premiers articles
df = df.sort_values(by='count', ascending=False).head(5)

# Créer le graphique avec Plotly
fig = px.bar(df, x=df.index, y='count', labels={'x': 'Articles', 'count': 'Nombre de commandes'},
             title="Articles les plus commandés chez Chipotle")
fig.show()
```

### Étape 6. Créer un nuage de points avec le nombre d'articles commandés par prix de commande
#### Indice : Le prix doit être sur l'axe X et les articles commandés sur l'axe Y

```python
# Convertir les prix en float (enlevant le signe dollar et les espaces)
chipo['item_price'] = chipo['item_price'].str.replace('$', '').astype(float)

# Grouper par commande et sommer les quantités et les prix
orders = chipo.groupby('order_id').sum()

# Créer le nuage de points avec Plotly
fig = px.scatter(orders, x='item_price', y='quantity',
                 labels={'item_price': 'Prix de la commande', 'quantity': 'Articles commandés'},
                 title="Nombre d'articles commandés par prix de commande",
                 size_max=10, color='quantity')
fig.show()
```

### BONUS : Créer une question et un graphique pour répondre à votre propre question.

```python
# Exemple de question : Quel est le prix moyen des articles commandés ?
avg_price_per_item = chipo.groupby('item_name')['item_price'].mean().sort_values(ascending=False)

# Créer un graphique pour les 10 articles les plus chers
df_avg_price = avg_price_per_item.head(10).reset_index()
fig = px.bar(df_avg_price, x='item_name', y='item_price',
             labels={'item_name': 'Articles', 'item_price': 'Prix moyen ($)'},
             title="Prix moyen des 10 articles les plus chers")
fig.show()
```

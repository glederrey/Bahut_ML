### Étape 1. Importez les bibliothèques nécessaires

```python
import pandas as pd
import datetime
```

### Étape 2. Importez le jeu de données depuis cette [adresse](https://github.com/guipsamora/pandas_exercises/blob/master/06_Stats/Wind_Stats/wind.data)

### Étape 3. Assignez-le à une variable appelée data et remplacez les 3 premières colonnes par un index datetime approprié.

```python
# parse_dates récupère les colonnes 0, 1, 2 et les analyse comme index
data_url = 'https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/06_Stats/Wind_Stats/wind.data'
data = pd.read_csv(data_url, sep = "\s+", parse_dates = [[0,1,2]]) 
data.head()
```

### Étape 4. Année 2061 ? Avons-nous vraiment des données de cette année ? Créez une fonction pour corriger cela et appliquez-la.

```python
# Le problème est que les dates sont en 2061 et ainsi de suite...

# fonction qui utilise datetime
def fix_century(x):
  year = x.year - 100 if x.year > 1989 else x.year
  return datetime.date(year, x.month, x.day)

# appliquez la fonction fix_century sur la colonne et remplacez les valeurs par les bonnes
data['Yr_Mo_Dy'] = data['Yr_Mo_Dy'].apply(fix_century)

# data.info()
data.head()
```

### Étape 5. Définissez les bonnes dates comme index. Faites attention au type de données, il doit être datetime64[ns].

```python
# transformez Yr_Mo_Dy en type date datetime64
data["Yr_Mo_Dy"] = pd.to_datetime(data["Yr_Mo_Dy"])

# définissez 'Yr_Mo_Dy' comme index
data = data.set_index('Yr_Mo_Dy')

data.head()
# data.info()
```

### Étape 10. Créez un DataFrame appelé day_stats et calculez les vitesses minimales, maximales et moyennes du vent ainsi que les écarts-types des vitesses du vent sur tous les emplacements chaque jour.

#### Un ensemble différent de nombres pour chaque jour.

```python
# créez le DataFrame
day_stats = pd.DataFrame()

# cette fois, nous déterminons axis égal à un pour qu'il prenne chaque ligne.
day_stats['min'] = data.min(axis = 1) # min
day_stats['max'] = data.max(axis = 1) # max 
day_stats['mean'] = data.mean(axis = 1) # moyenne
day_stats['std'] = data.std(axis = 1) # écarts-types

day_stats.head()
```

### Étape 15. Calculez les vitesses minimales, maximales et moyennes du vent ainsi que les écarts-types des vitesses du vent sur tous les emplacements pour chaque semaine (supposez que la première semaine commence le 2 janvier 1961) pour les 52 premières semaines.

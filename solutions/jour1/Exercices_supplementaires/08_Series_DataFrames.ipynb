### Étape 1. Importer les bibliothèques nécessaires

```python
import pandas as pd
```

### Étape 2. Créer un dictionnaire de données

```python
raw_data = {
    "name": ['Bulbizarre', 'Salamèche', 'Carapuce', 'Chenipan'],
    "evolution": ['Herbizarre', 'Reptincel', 'Carabaffe', 'Chrysacier'],
    "type": ['plante', 'feu', 'eau', 'insecte'],
    "hp": [45, 39, 44, 45],
    "pokedex": ['oui', 'non', 'oui', 'non']
}
```

### Étape 3. L'assigner à une variable appelée `pokemon`

```python
pokemon = pd.DataFrame(raw_data)
pokemon.head()
```

### Étape 4. Oups... il semble que les colonnes du DataFrame soient dans l'ordre alphabétique. Réorganisez-les dans l'ordre suivant : `name`, `type`, `hp`, `evolution`, `pokedex`.

```python
pokemon = pokemon[['name', 'type', 'hp', 'evolution', 'pokedex']]
pokemon
```

### Étape 5. Ajouter une autre colonne appelée `place` et insérer ce que vous avez en tête.

```python
pokemon['place'] = ['parc', 'rue', 'lac', 'forêt']
pokemon
```

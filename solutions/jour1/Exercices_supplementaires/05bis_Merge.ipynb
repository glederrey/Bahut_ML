### Étape 1. Importez les bibliothèques nécessaires

```python
import pandas as pd
```

### Étape 2. Créez les 3 DataFrames à partir des données brutes suivantes

```python
raw_data_1 = {
        'subject_id': ['1', '2', '3', '4', '5'],
        'first_name': ['Alex', 'Amy', 'Allen', 'Alice', 'Ayoung'], 
        'last_name': ['Anderson', 'Ackerman', 'Ali', 'Aoni', 'Atiches']}

raw_data_2 = {
        'subject_id': ['4', '5', '6', '7', '8'],
        'first_name': ['Billy', 'Brian', 'Bran', 'Bryce', 'Betty'], 
        'last_name': ['Bonder', 'Black', 'Balwner', 'Brice', 'Btisan']}

raw_data_3 = {
        'subject_id': ['1', '2', '3', '4', '5', '7', '8', '9', '10', '11'],
        'test_id': [51, 15, 15, 61, 16, 14, 15, 1, 61, 16]}
```

### Étape 3. Assignez-les à des variables appelées data1, data2, data3

```python
data1 = pd.DataFrame(raw_data_1, columns = ['subject_id', 'first_name', 'last_name'])
data2 = pd.DataFrame(raw_data_2, columns = ['subject_id', 'first_name', 'last_name'])
data3 = pd.DataFrame(raw_data_3, columns = ['subject_id','test_id'])

data3
```

### Étape 4. Fusionnez les deux DataFrames par lignes et assignez-les à all_data

```python
all_data = pd.concat([data1, data2])
all_data
```

### Étape 5. Fusionnez les deux DataFrames par colonnes et assignez-les à all_data_col

```python
all_data_col = pd.concat([data1, data2], axis = 1)
all_data_col
```

### Étape 6. Affichez data3

```python
data3
```

### Étape 7. Fusionnez all_data et data3 en fonction de la valeur subject_id

```python
pd.merge(all_data, data3, on='subject_id')
```

### Étape 8. Fusionnez uniquement les données ayant le même 'subject_id' dans data1 et data2

```python
pd.merge(data1, data2, on='subject_id', how='inner')
```

### Étape 9. Fusionnez toutes les valeurs de data1 et data2, avec les enregistrements correspondants des deux côtés lorsque disponibles.

```python
pd.merge(data1, data2, on='subject_id', how='outer')
```

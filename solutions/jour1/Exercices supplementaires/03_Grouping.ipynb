### Étape 1. Importez les bibliothèques nécessaires

```python
import pandas as pd
```

### Étape 2. Importez le jeu de données depuis cette [adresse](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/drinks.csv).

### Étape 3. Assignez-le à une variable appelée drinks.

```python
drinks = pd.read_csv('https://raw.githubusercontent.com/justmarkham/DAT8/master/data/drinks.csv')
drinks.head()
```

### Étape 4. Quel continent consomme le plus de bière en moyenne ?

```python
drinks.groupby('continent').beer_servings.mean()
```

### Étape 5. Pour chaque continent, affichez les statistiques de la consommation de vin.

```python
drinks.groupby('continent').wine_servings.describe()
```

### Étape 6. Affichez la consommation moyenne d'alcool par continent pour chaque colonne

```python
numeric_columns = ['beer_servings', 'spirit_servings', 'wine_servings', 'total_litres_of_pure_alcohol']
drinks.groupby('continent')[numeric_columns].mean()
```

### Étape 7. Affichez la consommation médiane d'alcool par continent pour chaque colonne

```python
drinks.groupby('continent')[numeric_columns].median()
```

### Étape 8. Affichez les valeurs moyennes, minimales et maximales pour la consommation de spiritueux.
#### Cette fois, affichez un DataFrame

```python
drinks.groupby('continent').spirit_servings.agg(['mean', 'min', 'max'])
```

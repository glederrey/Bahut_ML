### Étape 1. Importer les bibliothèques nécessaires :

```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
```

### Étape 2. Importer le dataset depuis cette [adresse](https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/07_Visualization/Tips/tips.csv).

### Étape 3. L'assigner à une variable appelée `tips`.

```python
url = 'https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/07_Visualization/Tips/tips.csv'
tips = pd.read_csv(url)
tips.head()
```

### Étape 4. Supprimer la colonne `Unnamed: 0`

```python
tips = tips.drop(columns=['Unnamed: 0'])
tips.head()
```

### Étape 5. Tracer l'histogramme de la colonne `total_bill`

```python
fig = px.histogram(tips, x='total_bill', nbins=20, title='Histogramme des factures totales')
fig.update_layout(xaxis_title='Valeur', yaxis_title='Fréquence')
fig.show()
```

### Étape 6. Créer un nuage de points montrant la relation entre `total_bill` et `tip`

```python
fig = px.scatter(tips, x='total_bill', y='tip', title='Relation entre Total Bill et Tip',
                 labels={'total_bill': 'Facture totale', 'tip': 'Pourboire'})
fig.show()
```

### Étape 7. Créer une image montrant la relation entre `total_bill`, `tip` et `size`

```python
fig = px.scatter_3d(tips, x='total_bill', y='tip', z='size', color='size',
                    title='Relation entre Total Bill, Tip et Taille',
                    labels={'total_bill': 'Facture totale', 'tip': 'Pourboire', 'size': 'Taille'})
fig.show()
```

### Étape 8. Présenter la relation entre les jours et la valeur de `total_bill`

```python
fig = px.box(tips, x='day', y='total_bill', title='Relation entre les jours et la facture totale',
             labels={'day': 'Jour', 'total_bill': 'Facture totale'})
fig.show()
```

### Étape 9. Créer un nuage de points avec `day` sur l'axe Y et `tip` sur l'axe X, en différenciant les points par sexe

```python
fig = px.strip(tips, x='tip', y='day', color='sex', title='Pourboires par jour et par sexe',
               labels={'tip': 'Pourboire', 'day': 'Jour', 'sex': 'Sexe'})
fig.show()
```

### Étape 10. Créer un diagramme en boîte montrant `total_bill` par jour en différenciant le moment (Dîner ou Déjeuner)

```python
fig = px.box(tips, x='day', y='total_bill', color='time',
             title='Facture totale par jour et par moment',
             labels={'day': 'Jour', 'total_bill': 'Facture totale', 'time': 'Moment'})
fig.show()
```

### Étape 11. Créer deux histogrammes des valeurs de `tip` pour le dîner et le déjeuner, côte à côte

```python
fig = px.histogram(tips, x='tip', color='time', barmode='overlay',
                   title='Histogramme des pourboires pour le dîner et le déjeuner',
                   labels={'tip': 'Pourboire', 'time': 'Moment'})
fig.show()
```

### Étape 12. Créer deux nuages de points, un pour les hommes et un pour les femmes, montrant la relation entre `total_bill` et `tip`, en différenciant par fumeur ou non-fumeur

```python
fig = px.scatter(tips, x='total_bill', y='tip', color='smoker', facet_col='sex',
                 title='Relation entre Facture Totale et Pourboire par Sexe et Fumeur',
                 labels={'total_bill': 'Facture totale', 'tip': 'Pourboire', 'smoker': 'Fumeur'})
fig.show()
```

L'exercice suivant utilise des données provenant du [UC Irvine Machine Learning Repository](https://archive.ics.uci.edu/ml/datasets/Auto+MPG).

### Étape 1. Importez les bibliothèques nécessaires

```python
import pandas as pd
import numpy as np
```

### Étape 2. Importez le premier jeu de données [cars1](https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/05_Merge/Auto_MPG/cars1.csv) et [cars2](https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/05_Merge/Auto_MPG/cars2.csv).

   ### Étape 3. Assignez-les à des variables appelées cars1 et cars2

```python
cars1 = pd.read_csv("https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/05_Merge/Auto_MPG/cars1.csv")
cars2 = pd.read_csv("https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/05_Merge/Auto_MPG/cars2.csv")

print(cars1.head())
print(cars2.head())
```

### Étape 4. Oups, il semble que notre premier jeu de données ait des colonnes vides non nommées. Corrigez cars1.

```python
cars1 = cars1.loc[:, "mpg":"car"]
cars1.head()
```

### Étape 5. Quel est le nombre d'observations dans chaque jeu de données ?

```python
print(cars1.shape)
print(cars2.shape)
```

### Étape 6. Fusionnez cars1 et cars2 en un seul DataFrame appelé cars.

```python
cars = pd.concat([cars1, cars2])
cars
```

### Étape 7. Oups, il manque une colonne appelée owners. Créez une série de nombres aléatoires entre 15 000 et 73 000.

```python
nr_owners = np.random.randint(15000, high=73001, size=398, dtype='l')
nr_owners
```

### Étape 8. Ajoutez la colonne owners à cars.

```python
cars['owners'] = nr_owners
cars.tail()
```

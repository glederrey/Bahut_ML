### Évaluation du modèle de classification

Nous avons essayé de trouver les meilleurs hyperparamètres sur notre modèle en utilisant `RandomizedSearchCV` et jusqu'à présent, nous n'avons évalué notre modèle qu'en utilisant la fonction `score()` qui renvoie la précision. 

Mais en ce qui concerne la classification, vous voudrez probablement utiliser quelques métriques d'évaluation supplémentaires, notamment :
* [**Matrice de confusion**](https://www.dataschool.io/simple-guide-to-confusion-matrix-terminology/) - Compare les valeurs prédites avec les valeurs réelles de manière tabulaire, si 100% correct, toutes les valeurs dans la matrice seront de haut en bas à gauche à droite (ligne diagonale).
* [**Validation croisée**](https://scikit-learn.org/stable/modules/cross_validation.html) - Divise votre jeu de données en plusieurs parties et entraîne et teste votre modèle sur chaque partie et évalue les performances comme une moyenne. 
* [**Précision**](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.precision_score.html#sklearn.metrics.precision_score) - Proportion de vrais positifs sur le nombre total d'échantillons. Une précision plus élevée conduit à moins de faux positifs.
* [**Rappel**](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.recall_score.html#sklearn.metrics.recall_score) - Proportion de vrais positifs sur le nombre total de vrais positifs et de faux positifs. Un rappel plus élevé conduit à moins de faux négatifs.
* [**Score F1**](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.f1_score.html#sklearn.metrics.f1_score) - Combine la précision et le rappel en une seule métrique. 1 est le meilleur, 0 est le pire.
* [**Rapport de classification**](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.classification_report.html) - Sklearn a une fonction intégrée appelée `classification_report()` qui renvoie certaines des principales métriques de classification telles que la précision, le rappel et le score F1.
* [**Courbe ROC**](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.roc_score.html) - [Receiver Operating Characteristic](https://en.wikipedia.org/wiki/Receiver_operating_characteristic) est un tracé du taux de vrais positifs par rapport au taux de faux positifs.
* [**Aire sous la courbe (AUC)**](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.roc_auc_score.html) - L'aire sous la courbe ROC. Un modèle parfait atteint un score de 1.0.

**Défi :** Notez ce que représente chacune des colonnes de ce rapport de classification.

* **Précision** - Indique la proportion d'identifications positives (classe 1 prédite par le modèle) qui étaient réellement correctes. Un modèle qui ne produit aucun faux positif a une précision de 1.0.
* **Rappel** - Indique la proportion de positifs réels qui ont été correctement classés. Un modèle qui ne produit aucun faux négatif a un rappel de 1.0.
* **Score F1** - Une combinaison de précision et de rappel. Un modèle parfait atteint un score F1 de 1.0.
* **Support** - Le nombre d'échantillons sur lesquels chaque métrique a été calculée.
* **Précision** - La précision du modèle sous forme décimale. Une précision parfaite est égale à 1.0.
* **Macro avg** - Abréviation de moyenne macro, la moyenne de précision, rappel et score F1 entre les classes. La moyenne macro ne prend pas en compte le déséquilibre des classes, donc si vous avez des déséquilibres de classes, faites attention à cette métrique.
* **Weighted avg** - Abréviation de moyenne pondérée, la moyenne pondérée de précision, rappel et score F1 entre les classes. Pondéré signifie que chaque métrique est calculée en tenant compte du nombre d'échantillons dans chaque classe. Cette métrique favorisera la classe majoritaire (par exemple, elle donnera une valeur élevée lorsqu'une classe surpasse une autre en raison d'un plus grand nombre d'échantillons).

Puisque nous avons ajusté certains modèles mais ne les avons comparés que via la métrique par défaut contenue dans la méthode `score()` (score R^2 ou coefficient de détermination), prenons le modèle `RidgeRegression` et évaluons-le avec quelques autres [métriques de régression](https://scikit-learn.org/stable/modules/model_evaluation.html#regression-metrics).

Plus précisément, trouvons :
1. **R^2 (prononcé r-carré) ou coefficient de détermination** - Compare les prédictions de votre modèle à la moyenne des cibles. Les valeurs peuvent aller de moins l'infini (un très mauvais modèle) à 1. Par exemple, si tout ce que fait votre modèle est de prédire la moyenne des cibles, sa valeur R^2 serait de 0. Et si votre modèle prédit parfaitement une gamme de nombres, sa valeur R^2 serait de 1. 
2. **Erreur absolue moyenne (MAE)** - La moyenne des différences absolues entre les prédictions et les valeurs réelles. Cela vous donne une idée de l'ampleur des erreurs de vos prédictions.
3. **Erreur quadratique moyenne (MSE)** - La moyenne des différences au carré entre les prédictions et les valeurs réelles. Élever les erreurs au carré élimine les erreurs négatives. Cela amplifie également les valeurs aberrantes (échantillons qui ont des erreurs plus importantes).

Scikit-Learn dispose de quelques classes intégrées qui vont nous aider avec celles-ci, à savoir, [`mean_absolute_error`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.mean_absolute_error.html), [`mean_squared_error`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.mean_squared_error.html) et [`r2_score`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.r2_score.html).

### Flux de travail du classificateur Random Forest pour la classification des maladies cardiaques

#### 1. Préparer les données

Comme exemple de jeu de données, nous allons importer `heart-disease.csv`.

Ce fichier contient des dossiers médicaux anonymisés de patients et indique s'ils souffrent ou non d'une maladie cardiaque (il s'agit d'un problème de classification puisque nous essayons de prédire si quelque chose est une chose ou une autre).

````python
import pandas as pd

# heart_disease = pd.read_csv("../data/heart-disease.csv") # charger les données depuis le répertoire local
heart_disease = pd.read_csv("https://raw.githubusercontent.com/mrdbourke/zero-to-mastery-ml/master/data/heart-disease.csv") # charger les données directement depuis l'URL (nécessite la forme brute sur GitHub, source : https://github.com/mrdbourke/zero-to-mastery-ml/blob/master/data/heart-disease.csv)
heart_disease.head()
````

Ici, chaque ligne représente un patient différent et toutes les colonnes sauf `target` sont différentes caractéristiques du patient. 

La colonne `target` indique si le patient souffre d'une maladie cardiaque (`target=1`) ou non (`target=0`), c'est notre colonne "étiquette", la variable que nous allons essayer de prédire.

Le reste des colonnes (souvent appelées caractéristiques ou features) sont ce que nous utiliserons pour prédire la valeur `target`.

> **Remarque :** Il est courant de sauvegarder les caractéristiques dans une variable `X` et les étiquettes dans une variable `y`. En pratique, nous aimerions utiliser les `X` (caractéristiques) pour construire un algorithme prédictif afin de prédire les `y` (étiquettes).

````python
# Créer X (toutes les colonnes de caractéristiques)
X = heart_disease.drop("target", axis=1)

# Créer y (la colonne cible)
y = heart_disease["target"]

# Vérifier les premières lignes du DataFrame de caractéristiques
X.head()
````

````python
# Vérifier les premières lignes et le décompte des valeurs des étiquettes
y.head(), y.value_counts()
````

L'une des pratiques les plus importantes en machine learning est de diviser les jeux de données en ensembles d'entraînement et de test.

C'est-à-dire qu'un modèle va **s'entraîner sur l'ensemble d'entraînement** pour apprendre des motifs, puis ces motifs pourront être **évalués sur l'ensemble de test**.

Crucialmente, un modèle ne devrait **jamais** voir les données de test pendant l'entraînement.

C'est équivalent à un étudiant qui étudie les matériaux du cours pendant le semestre (ensemble d'entraînement) puis teste ses capacités lors de l'examen suivant (ensemble de test).

Scikit-learn fournit la méthode [`sklearn.model_selection.train_test_split`](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.train_test_split.html) pour diviser les jeux de données en ensembles d'entraînement et de test.

> **Remarque :** Une pratique courante consiste à utiliser une répartition 80/20 ou 70/30 ou 75/25 pour les données d'entraînement/test. Il existe également un troisième ensemble, connu sous le nom d'ensemble de validation (par exemple 70/15/15 pour entraînement/validation/test) pour l'ajustement des hyperparamètres, mais pour l'instant, nous nous concentrerons sur les ensembles d'entraînement et de test.

````python
# Diviser les données en ensembles d'entraînement et de test
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, 
                                                    y,
                                                    test_size=0.25) # par défaut train_test_split utilise 25% des données pour l'ensemble de test

X_train.shape, X_test.shape, y_train.shape, y_test.shape
````

#### 2. Choisir le modèle et les hyperparamètres

Le choix d'un modèle dépend souvent du type de problème sur lequel vous travaillez.

Par exemple, Scikit-Learn recommande différents modèles selon que vous travaillez sur un problème de classification ou de régression.

Vous pouvez consulter une carte détaillant les [différents types d'options de modèles et les recommandations dans la documentation Scikit-Learn](https://scikit-learn.org/stable/machine_learning_map.html).

Scikit-Learn fait référence aux modèles en tant qu'"estimateurs", cependant, ils sont souvent également appelés `model` ou `clf` (abréviation de classifier).

Les hyperparamètres d'un modèle sont des paramètres que vous pouvez modifier pour l'adapter à votre problème, un peu comme les boutons d'un four que vous pouvez régler pour cuisiner votre plat préféré.

````python
# Puisque nous travaillons sur un problème de classification, nous allons commencer avec un RandomForestClassifier
from sklearn.ensemble import RandomForestClassifier

clf = RandomForestClassifier()
````

Nous pouvons voir les hyperparamètres actuels d'un modèle avec la méthode [`get_params()`](https://scikit-learn.org/stable/developers/develop.html#get-params-and-set-params).

````python
# Voir les hyperparamètres actuels
clf.get_params()
````

Nous allons laisser cela tel quel pour l'instant, car les modèles Scikit-Learn ont généralement de bons paramètres par défaut.

#### 3. Ajuster le modèle aux données et l'utiliser pour faire une prédiction

Ajuster un modèle à un jeu de données consiste à lui transmettre les données et à lui demander de déterminer les motifs.

S'il y a des étiquettes (apprentissage supervisé), le modèle essaie de déterminer la relation entre les données et les étiquettes. 

S'il n'y a pas d'étiquettes (apprentissage non supervisé), le modèle essaie de trouver des motifs et de regrouper des échantillons similaires.

La plupart des modèles Scikit-Learn ont la méthode [`fit(X, y)`](https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.RandomForestClassifier.html#sklearn.ensemble.RandomForestClassifier.fit) intégrée, où le paramètre `X` représente les caractéristiques et le paramètre `y` représente les étiquettes.

Dans notre cas, nous commençons par ajuster un modèle sur la division d'entraînement (`X_train`, `y_train`).

````python
clf.fit(X=X_train, y=y_train)
````

#### Utiliser le modèle pour faire une prédiction

L'objectif principal de l'entraînement d'un modèle de machine learning est de l'utiliser pour faire une sorte de prédiction à l'avenir.

Une fois que votre instance de modèle est entraînée, vous pouvez utiliser la méthode [`predict()`](https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.RandomForestClassifier.html#sklearn.ensemble.RandomForestClassifier.predict) pour prédire une valeur cible étant donné un ensemble de caractéristiques. 

En d'autres termes, utilisez le modèle, avec de nouvelles données non vues et non étiquetées pour prédire l'étiquette.

> **Remarque :** Les données sur lesquelles vous faites des prédictions doivent avoir la même forme et le même format que les données sur lesquelles vous avez entraîné le modèle.

````python
# Cela ne fonctionne pas... formes incorrectes
y_label = clf.predict(np.array([0, 2, 3, 4]))
````

Oh non ! 

Nous obtenons une `ValueError` (formes incompatibles) :

```
ValueError: Expected 2D array, got 1D array instead:
array=[0. 2. 3. 4.].
Reshape your data either using array.reshape(-1, 1) if your data has a single feature or array.reshape(1, -1) if it contains a single sample.
```

Cela se produit parce que nous **essayons de faire des prédictions sur des données qui sont dans un format différent de celui des données sur lesquelles notre modèle a été entraîné**.

Puisque notre modèle a été entraîné sur les données de `X_train`, les prédictions doivent être faites sur des données ayant le même format et la même forme que `X_train`.

Notre objectif dans de nombreux problèmes de machine learning est d'utiliser les motifs appris à partir des données d'entraînement pour faire des prédictions sur les données de test (ou des données futures non vues).

````python
# Pour prédire une étiquette, les données doivent avoir la même forme que X_train
X_test.head()
````

````python
# Utiliser le modèle pour faire une prédiction sur les données de test (évaluation plus approfondie)
y_preds = clf.predict(X=X_test)
````

#### 4. Évaluer le modèle

Maintenant que nous avons fait quelques prédictions, nous pouvons commencer à utiliser d'autres méthodes Scikit-Learn pour déterminer la qualité de notre modèle. 

Chaque modèle ou estimateur possède une méthode [`score()`](https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.RandomForestClassifier.html#sklearn.ensemble.RandomForestClassifier.score) intégrée. 

Cette méthode compare à quel point le modèle a pu apprendre les motifs entre les caractéristiques et les étiquettes.

La méthode `score()` pour chaque modèle utilise une métrique d'évaluation standard pour mesurer les résultats de votre modèle.

Dans le cas d'un classificateur (notre modèle), l'une des métriques d'évaluation les plus courantes est la [précision](https://scikit-learn.org/stable/modules/model_evaluation.html#accuracy-score) (la fraction de prédictions correctes sur le total des prédictions).

Vérifions la précision de notre modèle sur l'ensemble d'entraînement.

````python
# Évaluer le modèle sur l'ensemble d'entraînement
train_acc = clf.score(X=X_train, y=y_train)
print(f"La précision du modèle sur l'ensemble de données d'entraînement est : {train_acc*100}%")
````

Woah ! Il semble que notre modèle fonctionne assez bien sur l'ensemble de données d'entraînement.

C'est parce qu'il a eu la chance de voir à la fois les données *et* les étiquettes.

Qu'en est-il de l'ensemble de données de test ?

````python
# Évaluer le modèle sur l'ensemble de test
test_acc = clf.score(X=X_test, y=y_test)
print(f"La précision du modèle sur l'ensemble de données de test est : {test_acc*100:.2f}%")
````

Hmm, il semble que la précision de notre modèle soit un peu moins bonne sur l'ensemble de données de test que sur l'ensemble de données d'entraînement.

C'est assez souvent le cas, car rappelez-vous, un modèle n'a jamais vu les exemples de test auparavant.

Il existe également un certain nombre d'autres méthodes d'évaluation que nous pouvons utiliser pour nos modèles de classification.

Toutes les métriques de classification suivantes proviennent du module [`sklearn.metrics`](https://scikit-learn.org/stable/modules/model_evaluation.html#classification-metrics) :
* [`classification_report(y_true, y_pred)`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.classification_report.html#sklearn.metrics.classification_report) - Construit un rapport textuel montrant diverses métriques de classification telles que [précision, rappel](https://scikit-learn.org/stable/auto_examples/model_selection/plot_precision_recall.html) et score F1.
* [`confusion_matrix(y_true, y_pred)`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.confusion_matrix.html#sklearn.metrics.confusion_matrix) - Crée une [matrice de confusion](https://en.wikipedia.org/wiki/Confusion_matrix) pour comparer les prédictions aux étiquettes de vérité.
* [`accuracy_score(y_true, y_pred)`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.accuracy_score.html#sklearn.metrics.accuracy_score) - Trouve le score de précision (la métrique par défaut) pour un classificateur.

Toutes les métriques ont en commun : elles comparent les prédictions d'un modèle (`y_pred`) aux étiquettes de vérité (`y_true`).

````python
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# Créer un rapport de classification
print(classification_report(y_test, y_preds))
````

````python
# Créer une matrice de confusion
conf_mat = confusion_matrix(y_test, y_preds)
conf_mat
````

````python
# Calculer le score de précision (identique à la méthode score() pour les classificateurs) 
accuracy_score(y_test, y_preds)
````

#### 5. Expérimenter pour améliorer

Le premier modèle que vous construisez est souvent appelé une référence (baseline) (une référence est souvent encore plus simple que le modèle que nous avons utilisé, une référence pourrait être "prédisons par défaut la valeur la plus courante et essayons ensuite de l'améliorer").

Une fois que vous avez un modèle de référence, comme celui que nous avons ici, il est important de se rappeler que ce n'est souvent pas le modèle final que vous utiliserez.

L'étape suivante du flux de travail consiste à essayer d'améliorer votre modèle de référence.

Comment ?

Avec l'une des devises les plus importantes en machine learning...

*Expérimenter, expérimenter, expérimenter !*

Les expériences peuvent prendre de nombreuses formes différentes.

Mais divisons-les en deux catégories.

1. Du point de vue du modèle.
2. Du point de vue des données.

Du point de vue du modèle, cela peut impliquer des choses comme l'utilisation d'un modèle plus complexe ou l'ajustement des hyperparamètres de vos modèles.

Du point de vue des données, cela peut impliquer la collecte de plus de données ou de données de meilleure qualité afin que votre modèle existant ait plus de chances d'apprendre les motifs qu'elles contiennent.

Si vous travaillez déjà sur un jeu de données existant, il est souvent plus facile d'essayer d'abord une série d'expériences du point de vue du modèle, puis de passer à des expériences du point de vue des données si vous n'obtenez pas les résultats que vous recherchez.

Une chose dont vous devez être conscient est que si vous ajustez les hyperparamètres d'un modèle dans une série d'expériences, vos résultats doivent toujours être validés par validation croisée (nous verrons cela plus tard !). 

La [validation croisée](http://scikit-learn.org/stable/modules/cross_validation.html) est un moyen de s'assurer que les résultats que vous obtenez sont cohérents entre vos jeux de données d'entraînement et de test (car elle utilise plusieurs versions de jeux d'entraînement et de test) plutôt que simplement dus à la chance en raison de l'ordre dans lequel les jeux d'entraînement et de test originaux ont été créés. 

* Essayez différents hyperparamètres.
* Tous les différents paramètres doivent être validés par validation croisée.
    * **Remarque :** Méfiez-vous de la validation croisée pour les problèmes de séries temporelles (car pour les séries temporelles, vous ne voulez pas mélanger des échantillons du futur avec des échantillons du passé).
    
Les différents modèles que vous utilisez auront différents hyperparamètres que vous pouvez ajuster. 

Dans le cas de notre modèle, le [`RandomForestClassifier()`](https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.RandomForestClassifier.html), nous allons commencer à essayer différentes valeurs pour `n_estimators` (une mesure du nombre d'arbres dans la forêt aléatoire).

Par défaut, `n_estimators=100`, alors essayons des valeurs de `100` à `200` et voyons ce qui se passe (généralement, plus c'est mieux) ?

````python
# Essayer différents nombres d'estimateurs (arbres)... (sans validation croisée)
np.random.seed(42)
for i in range(100, 200, 10):
    print(f"Essai du modèle avec {i} estimateurs...")
    model = RandomForestClassifier(n_estimators=i).fit(X_train, y_train)
    print(f"Précision du modèle sur l'ensemble de test : {model.score(X_test, y_test) * 100:.2f}%")
    print("")
````

Les métriques ci-dessus ont été mesurées sur une seule division d'entraînement et de test.

Utilisons [`sklearn.model_selection.cross_val_score`](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.cross_val_score.html) pour mesurer les résultats sur 5 jeux d'entraînement et de test différents.

Nous pouvons y parvenir en définissant `cross_val_score(X, y, cv=5)`.

Où `X` est l'ensemble *complet* des caractéristiques et `y` est l'ensemble *complet* des étiquettes et `cv` est le nombre de divisions d'entraînement et de test que `cross_val_score` créera automatiquement à partir des données (dans notre cas, `5` divisions différentes, c'est ce qu'on appelle la validation croisée à 5 plis).

````python
from sklearn.model_selection import cross_val_score

# Avec validation croisée
np.random.seed(42)
for i in range(100, 200, 10):
    print(f"Essai du modèle avec {i} estimateurs...")
    model = RandomForestClassifier(n_estimators=i).fit(X_train, y_train)

    # Mesurer le score du modèle sur une seule division entraînement/test
    model_score = model.score(X_test, y_test)
    print(f"Précision du modèle sur une seule division de l'ensemble de test : {model_score * 100:.2f}%")
    
    # Mesurer le score moyen de validation croisée sur 5 divisions différentes d'entraînement et de test
    cross_val_mean = np.mean(cross_val_score(model, X, y, cv=5))
    print(f"Score de validation croisée à 5 plis : {cross_val_mean * 100:.2f}%")
    
    print("")
````

Quel modèle a obtenu le meilleur score de validation croisée ?

C'est généralement un meilleur indicateur d'un modèle de qualité qu'un score de précision sur une seule division.

Plutôt que de configurer et de suivre manuellement les résultats de ces expériences, nous pouvons demander à Scikit-Learn de faire l'exploration pour nous.

Le [`sklearn.model_selection.GridSearchCV`](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.GridSearchCV.html) de Scikit-Learn est un moyen de rechercher un ensemble de différentes valeurs d'hyperparamètres et de suivre automatiquement celles qui fonctionnent le mieux.

Testons-le !

````python
# Une autre façon de le faire avec GridSearchCV...
np.random.seed(42)
from sklearn.model_selection import GridSearchCV

# Définir les paramètres à rechercher sous forme de dictionnaire 
# (il peut s'agir de n'importe lequel des hyperparamètres de votre modèle cible) 
param_grid = {'n_estimators': [i for i in range(100, 200, 10)]}

# Configurer la recherche par grille
grid = GridSearchCV(estimator=RandomForestClassifier(),
                    param_grid=param_grid,
                    cv=5,
                    verbose=1) 

# Ajuster la recherche par grille aux données
grid.fit(X, y)

# Trouver les meilleurs paramètres
print(f"Les meilleures valeurs de paramètres sont : {grid.best_params_}")
print(f"Avec un score de : {grid.best_score_*100:.2f}%")
````

Nous pouvons extraire le meilleur modèle/estimateur avec l'attribut `best_estimator_`.

````python
# Définir le modèle comme étant le meilleur estimateur
clf = grid.best_estimator_
clf
````

Et maintenant que nous avons le meilleur modèle validé par validation croisée, nous pouvons l'ajuster et l'évaluer sur notre division originale unique entraînement/test des données.

````python
# Ajuster le meilleur modèle
clf = clf.fit(X_train, y_train)

# Trouver les meilleurs scores de modèle sur notre division de test unique
# (remarque : cela peut être inférieur au score de validation croisée puisqu'il ne s'agit que d'une division des données)
print(f"Score du meilleur modèle sur une division unique des données : {clf.score(X_test, y_test)*100:.2f}%")
````

Les modèles de machine learning préfèrent travailler avec des nombres plutôt qu'avec du texte. 

Nous devrons donc d'abord convertir les caractéristiques non numériques en nombres.

Le processus de transformation des caractéristiques catégorielles en nombres est souvent appelé **encodage**.

Scikit-Learn propose un guide approfondi fantastique sur [*l'encodage des caractéristiques catégorielles*](https://scikit-learn.org/stable/modules/preprocessing.html#encoding-categorical-features).

Mais examinons l'une des façons les plus simples de transformer des caractéristiques catégorielles en nombres, l'encodage one-hot.

En machine learning, [l'encodage one-hot](https://en.wikipedia.org/wiki/One-hot#Machine_learning_and_statistics) donne une valeur de `1` à la valeur cible et une valeur de `0` aux autres valeurs.

Par exemple, supposons que nous ayons cinq échantillons et trois options de marque de voiture : Honda, Toyota, BMW.

Et nos échantillons étaient :
1. Honda
2. BMW
3. BMW
4. Toyota
5. Toyota

Si nous devions les encoder en one-hot, cela ressemblerait à :

| Échantillon | Honda | Toyota | BMW |
| ----- | ----- | ----- | ----- |
| 1 | 1 | 0 | 0 |
| 2 | 0 | 0 | 1 |
| 3 | 0 | 0 | 1 |
| 4 | 0 | 1 | 0 |
| 5 | 0 | 1 | 0 |

Remarquez comment il y a un 1 pour chaque valeur cible mais un 0 pour chaque autre valeur.

Nous pouvons utiliser les étapes suivantes pour encoder notre jeu de données en one-hot :
1. Importer [`sklearn.preprocessing.OneHotEncoder`](https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.OneHotEncoder.html) pour encoder nos caractéristiques en one-hot et [`sklearn.compose.ColumnTransformer`](https://scikit-learn.org/stable/modules/generated/sklearn.compose.ColumnTransformer.html) pour cibler les colonnes spécifiques de notre DataFrame à transformer.
2. Définir les caractéristiques catégorielles que nous aimerions transformer.
3. Créer une instance de `OneHotEncoder`.
4. Créer une instance de `ColumnTransformer` et lui fournir les transformations que nous aimerions effectuer.
5. Ajuster l'instance du `ColumnTransformer` à nos données et les transformer avec la méthode [`fit_transform(X)`](https://scikit-learn.org/stable/modules/generated/sklearn.compose.ColumnTransformer.html#sklearn.compose.ColumnTransformer.fit_transform).

> **Remarque :** Dans Scikit-Learn, le terme "transformer" est souvent utilisé pour désigner quelque chose qui *transforme* des données.

````python
# 1. Importer OneHotEncoder et ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

# 2. Définir les caractéristiques catégorielles à transformer
categorical_features = ["Make", "Colour", "Doors"]

# 3. Créer une instance de OneHotEncoder
one_hot = OneHotEncoder()

# 4. Créer une instance de ColumnTransformer
transformer = ColumnTransformer([("one_hot", # nom
                                  one_hot, # transformateur
                                  categorical_features)], # colonnes à transformer
                                  remainder="passthrough") # que faire avec le reste des colonnes ? ("passthrough" = laisser inchangé) 

# 5. Transformer les caractéristiques catégorielles en nombres (cela renverra une matrice creuse de type tableau, pas un DataFrame)
transformed_X = transformer.fit_transform(X)
transformed_X
````

> **Remarque :** Vous vous demandez peut-être pourquoi nous avons considéré `Doors` comme une variable catégorielle. C'est une bonne question étant donné que `Doors` est déjà numérique. Eh bien, la réponse est que `Doors` pourrait être soit numérique, soit catégorielle. Cependant, j'ai décidé de la considérer comme catégorielle, car là d'où je viens, le nombre de portes est souvent une *catégorie* différente de voiture. Par exemple, vous pouvez chercher des voitures à 4 portes ou des voitures à 5 portes (ce qui m'a toujours confondu car où est la 5ème porte ?). Cependant, vous pourriez expérimenter en traitant cette valeur comme numérique ou catégorielle, en entraînant un modèle sur chacune, puis en voyant comment chaque modèle se comporte.

### 1.2 Que faire s'il y a des valeurs manquantes dans les données ?

Des trous dans les données signifient des trous dans les motifs que votre modèle de machine learning peut apprendre.

De nombreux modèles de machine learning ne fonctionnent pas bien ou produisent des erreurs lorsqu'ils sont utilisés sur des jeux de données avec des valeurs manquantes.

Une valeur manquante peut apparaître comme un blanc, comme un `NaN` ou quelque chose de similaire.

Il existe deux options principales pour traiter les valeurs manquantes :

1. **Les remplir avec une valeur donnée ou calculée (imputation)** - Par exemple, vous pourriez remplir les valeurs manquantes d'une colonne numérique avec la moyenne de toutes les autres valeurs. La pratique consistant à calculer ou à déterminer comment remplir les valeurs manquantes dans un jeu de données est appelée **imputation**. Pour une excellente ressource sur l'imputation des valeurs manquantes, je vous recommande de consulter le [guide utilisateur de Scikit-Learn](https://scikit-learn.org/stable/modules/impute.html).
2. **Les supprimer** - Si une ligne ou un échantillon a des valeurs manquantes, vous pouvez choisir de les supprimer complètement de votre jeu de données. Cependant, cela peut potentiellement entraîner l'utilisation de moins de données pour construire votre modèle.

> **Remarque :** Le traitement des valeurs manquantes diffère d'un problème à l'autre, ce qui signifie qu'il n'existe pas de méthode 100 % optimale pour remplir les valeurs manquantes dans tous les jeux de données et types de problèmes. Il faudra souvent une expérimentation minutieuse et de la pratique pour déterminer la meilleure façon de traiter les valeurs manquantes dans vos propres jeux de données.

Pour nous entraîner à traiter les valeurs manquantes, importons une version du jeu de données `car_sales` avec plusieurs valeurs manquantes (à savoir [`car-sales-extended-missing-data.csv`](https://github.com/mrdbourke/zero-to-mastery-ml/blob/master/data/car-sales-extended-missing-data.csv)).

### 1.2.1 Remplir les données manquantes avec pandas

Voyons comment nous pourrions remplir les valeurs manquantes avec pandas.

Pour les valeurs catégorielles, l'une des façons les plus simples est de remplir les champs manquants avec la chaîne `"missing"`.

Nous pourrions faire cela pour les caractéristiques `Make` et `Colour`.

Quant à la caractéristique `Doors`, nous pourrions utiliser `"missing"` ou nous pourrions la remplir avec l'option la plus courante, soit `4`.

Avec la caractéristique `Odometer (KM)`, nous pouvons utiliser la valeur moyenne de toutes les autres valeurs de la colonne.

Et enfin, pour les échantillons qui n'ont pas de valeur `Price`, nous pouvons les supprimer (puisque `Price` est la valeur cible, la suppression cause probablement moins de dommages que l'imputation, cependant, vous pourriez concevoir une expérience pour tester cela).

| Colonne/Caractéristique | Remplir la valeur manquante avec | 
| ----- | ----- |
| `Make` | `"missing"` |
| `Colour` | `"missing"` |
| `Doors` | 4 (valeur la plus courante) |
| `Odometer (KM)` | moyenne de `Odometer (KM)` | 
| `Price` (cible) | NA, supprimer les échantillons sans `Price` |


> **Remarque :** La pratique consistant à remplir des données manquantes avec des valeurs données ou calculées est appelée [**imputation**](https://scikit-learn.org/stable/modules/impute.html). Et il est important de se rappeler qu'il n'existe pas de méthode parfaite pour remplir des données manquantes (à moins que ce ne soit avec des données qui auraient dû être là en premier lieu). Les méthodes que nous utilisons ne sont qu'une parmi tant d'autres. Les techniques que vous utilisez dépendront fortement de votre jeu de données. Un bon endroit pour chercher serait de rechercher "techniques d'imputation de données".

Commençons par la colonne `Make`.

Nous pouvons utiliser la méthode pandas [`fillna(value="missing", inplace=True)`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.fillna.html) pour remplir toutes les valeurs manquantes avec la chaîne `"missing"`.

````python
# Remplir les valeurs manquantes dans la colonne Make
# Remarque : Dans les versions précédentes de pandas, inplace=True était possible, mais cela sera modifié dans une future version, on peut utiliser la réaffectation à la place.
# car_sales_missing["Make"].fillna(value="missing", inplace=True)

car_sales_missing["Make"] = car_sales_missing["Make"].fillna(value="missing")
````

Et nous pouvons faire de même avec la colonne `Colour`.

````python
# Remarque : Dans les versions précédentes de pandas, inplace=True était possible, mais cela sera modifié dans une future version, on peut utiliser la réaffectation à la place.
# car_sales_missing["Colour"].fillna(value="missing", inplace=True)

# Remplir la colonne Colour
car_sales_missing["Colour"] = car_sales_missing["Colour"].fillna(value="missing")
````

````python
# Remplir la colonne Doors avec la valeur la plus courante
car_sales_missing["Doors"] = car_sales_missing["Doors"].fillna(value=4)
````

Nous pouvons supprimer les lignes avec des valeurs manquantes d'un DataFrame pandas avec la méthode [`pandas.DataFrame.dropna(inplace=True)`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.dropna.html).

Enfin, nous essaierons d'ajuster un modèle `RandomForestRegressor()` aux données nouvellement remplies.

### 1.2.2 Remplir les données manquantes et transformer les données catégorielles avec Scikit-Learn

Maintenant que nous avons rempli les colonnes manquantes à l'aide des fonctions pandas, vous vous demandez peut-être : "Pourquoi pandas ? Je pensais que c'était une introduction à Scikit-Learn ?".

Ne vous inquiétez pas, Scikit-Learn fournit une classe appelée [`sklearn.impute.SimpleImputer()`](https://scikit-learn.org/stable/modules/generated/sklearn.impute.SimpleImputer.html) qui nous permet de faire une chose similaire.

`SimpleImputer()` transforme les données en remplissant les valeurs manquantes avec une `strategy` donnée en paramètre.

Et nous pouvons l'utiliser pour remplir les valeurs manquantes dans notre DataFrame comme ci-dessus.

Pour l'instant, notre dataframe n'a pas de valeurs manquantes.

````python
car_sales_missing.isna().sum()
````

Réimportons-le pour qu'il ait des valeurs manquantes et que nous puissions les remplir avec Scikit-Learn.

````python
# Réimporter le DataFrame (pour que toutes les valeurs manquantes soient de retour)
# car_sales_missing = pd.read_csv("../data/car-sales-extended-missing-data.csv") # lecture depuis le répertoire local
car_sales_missing = pd.read_csv("https://raw.githubusercontent.com/mrdbourke/zero-to-mastery-ml/master/data/car-sales-extended-missing-data.csv") # lecture directement depuis l'URL (source: https://github.com/mrdbourke/zero-to-mastery-ml/blob/master/data/car-sales-extended-missing-data.csv)
car_sales_missing.isna().sum()
````

Pour commencer, nous allons supprimer les lignes qui n'ont pas de valeur dans la colonne `Price`.

````python
# Supprimer les lignes avec des valeurs manquantes dans la colonne Price
car_sales_missing.dropna(subset=["Price"], inplace=True)
````

Maintenant, il n'y a plus de lignes sans valeur `Price`.

````python
car_sales_missing.isna().sum()
````

Puisque nous n'avons pas à remplir de valeurs `Price`, divisons nos données en caractéristiques (`X`) et étiquettes (`y`).

Nous allons également diviser les données en ensembles d'entraînement et de test.

````python
# Diviser en X et y
X = car_sales_missing.drop("Price", axis=1)
y = car_sales_missing["Price"]

# Diviser les données en ensembles d'entraînement et de test
np.random.seed(42)
X_train, X_test, y_train, y_test = train_test_split(X,
                                                    y,
                                                    test_size=0.2)
````

> **Remarque :** Nous avons d'abord divisé les données en ensembles d'entraînement et de test pour traiter séparément les valeurs manquantes. C'est une bonne pratique car l'ensemble de test est censé simuler des données que le modèle n'a jamais vues auparavant. Pour les variables catégorielles, il est généralement acceptable de remplir les valeurs sur l'ensemble du jeu de données. Cependant, pour les variables numériques, vous devriez **remplir uniquement les valeurs de l'ensemble de test qui ont été calculées à partir de l'ensemble d'entraînement**.

Les ensembles d'entraînement et de test sont créés !

Configurons maintenant quelques instances de `SimpleImputer()` pour remplir diverses valeurs manquantes.

Nous utiliserons les stratégies et valeurs de remplissage suivantes :
* Pour les colonnes catégorielles (`Make`, `Colour`), `strategy="constant"`, `fill_value="missing"` (remplir les échantillons manquants avec une valeur constante de `"missing"`).
* Pour la colonne `Door`, `strategy="constant"`, `fill_value=4` (remplir les échantillons manquants avec une valeur constante de `4`).
* Pour la colonne numérique (`Odometer (KM)`), `strategy="mean"` (remplir les échantillons manquants avec la moyenne de la colonne cible).
  * **Remarque :** Il existe d'autres options de `strategy` et de remplissage dans la [documentation de `SimpleImputer()`](https://scikit-learn.org/stable/modules/generated/sklearn.impute.SimpleImputer.html).

````python
from sklearn.impute import SimpleImputer

# Créer un imputer pour les variables catégorielles
cat_imputer = SimpleImputer(strategy="constant", fill_value="missing")

# Créer un imputer pour la colonne Door
door_imputer = SimpleImputer(strategy="constant", fill_value=4)

# Créer un imputer pour la colonne Odometer (KM)
num_imputer = SimpleImputer(strategy="mean")
````

Les imputers sont créés !

Définissons maintenant les colonnes que nous souhaitons imputer.

Pourquoi ?

Parce que nous en aurons besoin sous peu (j'expliquerai dans la prochaine cellule de texte).

````python
# Définir différentes caractéristiques de colonnes
categorical_features = ["Make", "Colour"]
door_feature = ["Doors"]
numerical_feature = ["Odometer (KM)"]
````

Colonnes définies !

Maintenant, comment pourrions-nous transformer nos colonnes ?

Indice : nous pouvons utiliser la classe [`sklearn.compose.ColumnTransformer`](https://scikit-learn.org/stable/modules/generated/sklearn.compose.ColumnTransformer.html) de Scikit-Learn, d'une manière similaire à ce que nous avons fait précédemment pour convertir nos données en valeurs numériques.

C'est la raison pour laquelle nous avons défini les colonnes que nous souhaitons transformer.

Nous pouvons donc utiliser la classe `ColumnTransformer()`.

`ColumnTransformer()` prend en entrée une liste de tuples sous la forme `(nom_de_la_transformation, transformateur_à_utiliser, colonnes_à_transformer)` spécifiant quelles colonnes transformer et comment les transformer.

Par exemple :

```python
imputer = ColumnTransformer([
    ("cat_imputer", cat_imputer, categorical_features)
])
```

Dans ce cas, les variables du tuple sont :
* `nom_de_la_transformation` = `"cat_imputer"`
* `transformateur_à_utiliser` = `cat_imputer` (l'instance de `SimpleImputer()` que nous avons définie ci-dessus)
* `colonnes_à_transformer` = `categorical_features` (la liste des caractéristiques catégorielles que nous avons définie ci-dessus).

Développons cela en étendant l'exemple.

````python
from sklearn.compose import ColumnTransformer

# Créer une série de transformations de colonnes à effectuer
imputer = ColumnTransformer([
    ("cat_imputer", cat_imputer, categorical_features),
    ("door_imputer", door_imputer, door_feature),
    ("num_imputer", num_imputer, numerical_feature)])
````

Parfait !

L'étape suivante consiste à ajuster notre instance `ColumnTransformer()` (`imputer`) aux données d'entraînement et à transformer les données de test.

En d'autres termes, nous voulons :
1. Apprendre les valeurs d'imputation à partir de l'ensemble d'entraînement.
2. Remplir les valeurs manquantes dans l'ensemble d'entraînement avec les valeurs apprises en 1.
3. Remplir les valeurs manquantes dans l'ensemble de test avec les valeurs apprises en 1.

Pourquoi procéder ainsi ?

Dans notre cas, nous ne calculons pas beaucoup de variables (à l'exception de la moyenne de la colonne `Odometer (KM)`), cependant, rappelez-vous que l'ensemble de test doit toujours rester comme des données non vues.

Donc **lors du remplissage des valeurs dans l'ensemble de test, elles ne doivent être qu'avec des valeurs calculées ou imputées à partir de l'ensemble d'entraînement**.

Nous pouvons réaliser les étapes 1 et 2 simultanément avec la méthode [`ColumnTransformer.fit_transform()`](https://scikit-learn.org/stable/modules/generated/sklearn.compose.ColumnTransformer.html#sklearn.compose.ColumnTransformer.fit_transform) (`fit` = trouver les valeurs à remplir, `transform` = les remplir).

Ensuite, nous pouvons effectuer l'étape 3 avec la méthode [`ColumnTransformer.transform()`](https://scikit-learn.org/stable/modules/generated/sklearn.compose.ColumnTransformer.html#sklearn.compose.ColumnTransformer.transform) (nous voulons seulement transformer l'ensemble de test, pas apprendre différentes valeurs à remplir).

````python
# Trouver les valeurs à remplir et transformer les données d'entraînement
filled_X_train = imputer.fit_transform(X_train)

# Remplir les valeurs dans l'ensemble de test avec les valeurs apprises de l'ensemble d'entraînement
filled_X_test = imputer.transform(X_test)

# Vérifier filled_X_train
filled_X_train
````

Excellent !

Transformons maintenant nos tableaux `filled_X_train` et `filled_X_test` en DataFrames pour examiner leurs valeurs manquantes.

````python
# Reconvertir nos tableaux de données transformées en DataFrames
filled_X_train_df = pd.DataFrame(filled_X_train, 
                                 columns=["Make", "Colour", "Doors", "Odometer (KM)"])

filled_X_test_df = pd.DataFrame(filled_X_test, 
                                columns=["Make", "Colour", "Doors", "Odometer (KM)"])

# Vérifier les données manquantes dans l'ensemble d'entraînement
filled_X_train_df.isna().sum()
````

Y a-t-il des données manquantes dans l'ensemble de test ?

````python
# Vérifier les données manquantes dans l'ensemble de test
filled_X_test_df.isna().sum()
````

Et qu'en est-il de l'original ?

````python
# Vérifier l'original... il contient toujours des valeurs manquantes
car_sales_missing.isna().sum()
````

Parfait !

Plus de valeurs manquantes !

Mais attendez...

Est-ce que toutes nos données sont numériques ?

````python
filled_X_train_df.head()
````

Ah... il semble que nos colonnes `Make` et `Colour` soient toujours des chaînes de caractères.

Encodons-les en one-hot avec la colonne `Doors` pour nous assurer qu'elles sont numériques, comme nous l'avons fait précédemment.

````python
# Maintenant, encodons les caractéristiques en one-hot avec le même code qu'auparavant
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

categorical_features = ["Make", "Colour", "Doors"]

one_hot = OneHotEncoder()

transformer = ColumnTransformer([("one_hot", 
                                  one_hot, 
                                  categorical_features)],
                                remainder="passthrough",
                                sparse_threshold=0) # retourner une matrice creuse ou non

# Remplir les valeurs d'entraînement et de test séparément
transformed_X_train = transformer.fit_transform(filled_X_train_df)
transformed_X_test = transformer.transform(filled_X_test_df)

# Vérifier X_train transformé et rempli
transformed_X_train
````

Parfait !

Maintenant nos données sont :
1. Toutes numériques
2. Sans valeurs manquantes

Essayons d'ajuster un modèle !

````python
# Maintenant que nous avons transformé X, voyons si nous pouvons ajuster un modèle
np.random.seed(42)
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor()

# Assurez-vous d'utiliser les données transformées (données X remplies et encodées en one-hot)
model.fit(transformed_X_train, y_train)
model.score(transformed_X_test, y_test)
````

Vous avez peut-être remarqué que ce résultat est légèrement différent de celui obtenu précédemment.

Pourquoi pensez-vous que c'est le cas ?

C'est parce que nous avons créé nos ensembles d'entraînement et de test différemment.

Nous avons divisé les données en ensembles d'entraînement et de test *avant* de remplir les valeurs manquantes.

Précédemment, nous avons fait l'inverse, rempli les valeurs manquantes *avant* de diviser les données en ensembles d'entraînement et de test.

Cela peut conduire à une fuite d'informations de l'ensemble d'entraînement vers l'ensemble de test.

Rappelez-vous, l'un des concepts les plus importants en apprentissage automatique est de s'assurer que votre modèle ne voit *aucune* donnée de test avant l'évaluation.

Nous continuerons à pratiquer, mais pour l'instant, voici quelques-uns des principaux points à retenir :
* Gardez vos ensembles d'entraînement et de test séparés.
* La plupart des jeux de données que vous rencontrerez ne seront pas prêts à être utilisés immédiatement avec des modèles d'apprentissage automatique. Et certains peuvent nécessiter plus de préparation que d'autres pour être prêts à l'emploi.
* Pour la plupart des modèles d'apprentissage automatique, vos données doivent être numériques. Cela impliquera de convertir ce avec quoi vous travaillez en nombres. Ce processus est souvent appelé **ingénierie des caractéristiques** ou **encodage des caractéristiques**.
* Certains modèles d'apprentissage automatique ne sont pas compatibles avec les données manquantes. Le processus de remplissage des données manquantes est appelé **imputation de données**.

## 3. Ajuster le modèle aux données et l'utiliser pour faire des prédictions

Maintenant que vous avez choisi un modèle, l'étape suivante consiste à lui faire apprendre à partir des données afin qu'il puisse être utilisé pour des prédictions à l'avenir.

### 3.1 Ajuster un modèle aux données

Dans Scikit-Learn, le processus d'apprentissage des modèles à partir d'un jeu de données implique d'appeler la méthode `fit()` et de lui passer des données, comme `fit(X, y)`.

Où `X` est un tableau de caractéristiques et `y` est un tableau cible.

D'autres noms pour `X` incluent :
* Données
* Variables de caractéristiques
* Caractéristiques
D'autres noms pour `y` incluent :
* Étiquettes
* Variable cible

Pour l'apprentissage supervisé, il y a généralement un `X` et un `y`.

Pour l'apprentissage non supervisé, il n'y a pas de `y` (pas d'étiquettes).

Revenons à l'exemple de l'utilisation des données des patients (`X`) pour prédire s'ils ont une maladie cardiaque ou non (`y`).

````python
# Importer la classe de modèle RandomForestClassifier du module ensemble
from sklearn.ensemble import RandomForestClassifier

# Configurer la graine aléatoire
np.random.seed(42)

# Diviser les données en X (caractéristiques/données) et y (cible/étiquettes)
X = heart_disease.drop("target", axis=1)
y = heart_disease["target"]

# Diviser en ensembles d'entraînement et de test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Instancier le modèle (sur l'ensemble d'entraînement)
clf = RandomForestClassifier(n_estimators=100)

# Appeler la méthode fit sur le modèle et lui passer les données d'entraînement
clf.fit(X_train, y_train)

# Vérifier le score du modèle (sur l'ensemble de test)
clf.score(X_test, y_test)
````

Que se passe-t-il ici ?

L'appel de la méthode `fit()` amènera l'algorithme d'apprentissage automatique à tenter de trouver des modèles entre `X` et `y`. Ou s'il n'y a pas de `y`, il ne trouvera que les modèles au sein de `X`.

Voyons `X`.

````python
X.head()
````

Et `y`.

````python
y.head()
````

Passer `X` et `y` à `fit()` amènera le modèle à parcourir tous les exemples de `X` (données) et à voir quelle est leur `y` (étiquette) correspondante.

La façon dont le modèle fait cela est différente selon le modèle que vous utilisez.

Expliquer les détails de chacun prendrait un manuel entier.

Pour l'instant, vous pourriez l'imaginer similaire à la façon dont vous découvririez des modèles si vous aviez assez de temps.

Vous regarderiez les variables de caractéristiques, `X`, l'`age`, le `sex`, le `chol` (cholestérol) et verriez quelles différentes valeurs ont conduit aux étiquettes, `y`, `1` pour la maladie cardiaque, `0` pour l'absence de maladie cardiaque.

Ce concept, quel que soit le problème, est similaire dans tout l'apprentissage automatique.

**Pendant l'entraînement (trouver des modèles dans les données) :**

Un algorithme d'apprentissage automatique examine un jeu de données, trouve des modèles, essaie d'utiliser ces modèles pour prédire quelque chose et se corrige du mieux qu'il peut avec les données et les étiquettes disponibles. Il stocke ces modèles pour une utilisation ultérieure.

**Pendant les tests ou en production (utilisation des modèles appris) :**

Un algorithme d'apprentissage automatique utilise les modèles qu'il a précédemment appris dans un jeu de données pour faire une prédiction sur des données non vues.

### 3.2 Faire des prédictions à l'aide d'un modèle d'apprentissage automatique
Maintenant que nous avons un modèle entraîné, qui a espérons-le appris des modèles dans les données, vous voudrez l'utiliser pour faire des prédictions.

Scikit-Learn permet cela de plusieurs façons.

Deux des plus courantes et utiles sont [`predict()`](https://github.com/scikit-learn/scikit-learn/blob/5f3c3f037/sklearn/multiclass.py#L299) et [`predict_proba()`](https://github.com/scikit-learn/scikit-learn/blob/5f3c3f037/sklearn/linear_model/_logistic.py#L1617).

Voyons-les en action.

````python
# Utiliser un modèle entraîné pour faire des prédictions
clf.predict(X_test)
````

Étant donné des données sous la forme de `X`, la fonction `predict()` renvoie des étiquettes sous la forme de `y`.

> **Remarque :** Pour que la fonction `predict()` fonctionne, elle doit recevoir `X` (données) dans le même format que celui sur lequel le modèle a été entraîné. Par exemple, si un modèle a été entraîné sur 10 caractéristiques formatées d'une certaine manière, les prédictions doivent être faites sur des données avec 10 caractéristiques formatées de la même manière. Tout ce qui est différent et cela renverra une erreur.

Il est de pratique courante d'enregistrer ces prédictions dans une variable nommée quelque chose comme `y_preds` pour une comparaison ultérieure avec `y_test` ou `y_true` (généralement identique à `y_test`, juste un autre nom).

````python
# Comparer les prédictions à la vérité
y_preds = clf.predict(X_test)
np.mean(y_preds == y_test)
````

Une autre façon d'évaluer les prédictions (en les comparant aux étiquettes de vérité) est avec le module [`sklearn.metrics`](http://scikit-learn.org/stable/modules/model_evaluation.html) de Scikit-Learn.

À l'intérieur, vous trouverez des méthodes telles que [`accuracy_score()`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.accuracy_score.html), qui est la métrique d'évaluation par défaut pour les problèmes de classification.

````python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_preds)
````

`predict_proba()` renvoie les probabilités (proba est l'abréviation de probabilité) d'une étiquette de classification.

````python
# Renvoyer les probabilités plutôt que les étiquettes
clf.predict_proba(X_test[:5])
````

Voyons la différence.

````python
# Renvoyer les étiquettes
clf.predict(X_test[:5])
````

`predict_proba()` renvoie un tableau de cinq tableaux contenant chacun deux valeurs.

Chaque nombre est la probabilité d'une étiquette pour un échantillon donné.

````python
# Trouver les probabilités de prédiction pour 1 échantillon
clf.predict_proba(X_test[:1])
````

Cette sortie signifie que pour l'échantillon `X_test[:1]`, le modèle prédit l'étiquette 0 (indice 0) avec un score de probabilité de 0,9.

Comme le score de probabilité le plus élevé est à l'indice `0` (et qu'il est supérieur à 0,5), lors de l'utilisation de `predict()`, une étiquette de `0` est attribuée.

````python
# Renvoyer l'étiquette pour 1 échantillon
clf.predict(X_test[:1])
````

D'où vient 0,5 ?

Comme notre problème est une tâche de classification binaire (maladie cardiaque ou pas de maladie cardiaque), prédire une étiquette avec une probabilité de 0,5 à chaque fois serait la même chose qu'un lancer de pièce (deviner 50/50 à chaque fois).

Par conséquent, une fois que la probabilité de prédiction d'un échantillon dépasse 0,5 pour une certaine étiquette, cette étiquette lui est attribuée.

`predict()` peut également être utilisé pour les modèles de régression.

````python
# Importer la classe de modèle RandomForestRegressor du module ensemble
from sklearn.ensemble import RandomForestRegressor

# Configurer la graine aléatoire
np.random.seed(42)

# Diviser les données en caractéristiques (X) et étiquettes (y)
X = housing_df.drop("target", axis=1)
y = housing_df["target"]

# Diviser en ensembles d'entraînement et de test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Instancier et ajuster le modèle (sur l'ensemble d'entraînement)
model = RandomForestRegressor()
model.fit(X_train, y_train)

# Faire des prédictions
y_preds = model.predict(X_test)
````

Maintenant, nous pouvons évaluer notre modèle de régression en utilisant [`sklearn.metrics.mean_absolute_error`](http://scikit-learn.org/stable/modules/generated/sklearn.metrics.mean_absolute_error.html) qui renvoie l'erreur moyenne sur tous les échantillons.

````python
# Comparer les prédictions à la vérité
from sklearn.metrics import mean_absolute_error
mean_absolute_error(y_test, y_preds)
````

Maintenant que nous avons vu comment faire apprendre à un modèle à trouver des modèles dans les données en utilisant la fonction `fit()` et à faire des prédictions en utilisant ce qu'il a appris avec les fonctions `predict()` et `predict_proba()`, il est temps d'évaluer ces prédictions.

## 4. Évaluer un modèle

Une fois que vous avez entraîné un modèle, vous voudrez un moyen de mesurer la fiabilité de ses prédictions.

Dans l'ensemble, l'idée principale de l'évaluation d'un modèle est de **comparer les prédictions du modèle à ce qu'elles auraient dû être idéalement** (les étiquettes de vérité).

Scikit-Learn implémente 3 méthodes différentes d'évaluation des modèles.

1. La méthode `score()`. L'appel de `score()` sur une instance de modèle renverra une métrique associée au type de modèle que vous utilisez. La métrique dépend du modèle que vous utilisez.
2. Le paramètre `scoring`. Ce paramètre peut être passé à des méthodes telles que [`cross_val_score()`](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.cross_val_score.html#sklearn.model_selection.cross_val_score) ou [`GridSearchCV()`](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.GridSearchCV.html) pour indiquer à Scikit-Learn d'utiliser un type spécifique de métrique de score.
3. Les fonctions de métrique spécifiques au problème disponibles dans [`sklearn.metrics`](https://scikit-learn.org/stable/modules/classes.html#module-sklearn.metrics). Similaire à la façon dont le paramètre `scoring` peut recevoir différentes fonctions de score, Scikit-Learn les implémente en tant que fonctions autonomes.

La fonction de score que vous utilisez dépendra également du problème sur lequel vous travaillez.

Les problèmes de classification ont des métriques d'évaluation et des fonctions de score différentes des problèmes de régression.

Examinons quelques exemples.

### 4.1 Évaluation générale du modèle avec `score()`

Si nous reprenons le code de notre problème de classification précédent (construction d'un classificateur pour prédire si quelqu'un a une maladie cardiaque ou non en fonction de son dossier médical).

Nous pouvons voir la méthode `score()` entrer en jeu.

````python
# Importer la classe de modèle RandomForestClassifier du module ensemble
from sklearn.ensemble import RandomForestClassifier

# Configurer la graine aléatoire
np.random.seed(42)

# Diviser les données en X (caractéristiques/données) et y (cible/étiquettes)
X = heart_disease.drop("target", axis=1)
y = heart_disease["target"]

# Diviser en ensembles d'entraînement et de test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Instancier le modèle (sur l'ensemble d'entraînement)
clf = RandomForestClassifier(n_estimators=100)

# Appeler la méthode fit sur le modèle et lui passer les données d'entraînement
clf.fit(X_train, y_train);
````

Une fois que le modèle a été ajusté sur les données d'entraînement (`X_train`, `y_train`), nous pouvons appeler la méthode `score()` dessus et évaluer notre modèle sur les données de test, des données que le modèle n'a jamais vues auparavant (`X_test`, `y_test`).

````python
# Vérifier le score du modèle (sur l'ensemble de test)
clf.score(X_test, y_test)
````

Chaque modèle dans Scikit-Learn implémente une métrique par défaut pour `score()` qui est adaptée au problème.

Par exemple :
* Les modèles de classificateur utilisent généralement [`metrics.accuracy_score()`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.accuracy_score.html#sklearn.metrics.accuracy_score) comme métrique `score()` par défaut.

````python
# Importer cross_val_score du module model_selection
from sklearn.model_selection import cross_val_score

# Importer la classe de modèle RandomForestClassifier du module ensemble
from sklearn.ensemble import RandomForestClassifier

# Configurer la graine aléatoire
np.random.seed(42)

# Diviser les données en X (caractéristiques/données) et y (cible/étiquettes)
X = heart_disease.drop("target", axis=1)
y = heart_disease["target"]

# Diviser en ensembles d'entraînement et de test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Instancier le modèle (sur l'ensemble d'entraînement)
clf = RandomForestClassifier(n_estimators=100)

# Appeler la méthode fit sur le modèle et lui passer les données d'entraînement
clf.fit(X_train, y_train);
````

L'utilisation de `cross_val_score()` est légèrement différente de `score()`.

Voyons d'abord un exemple de code, puis nous examinerons les détails.

````python
# Utilisation de score()
clf.score(X_test, y_test)
````

````python
# Utilisation de cross_val_score()
cross_val_score(clf, X, y, cv=5) # cv = nombre de divisions à tester (5 par défaut)
````

Que se passe-t-il ici ?

La première différence que vous pourriez remarquer est que `cross_val_score()` renvoie un tableau alors que `score()` ne renvoie qu'un seul nombre.

`cross_val_score()` renvoie un tableau en raison d'un paramètre appelé `cv`, qui signifie validation croisée (cross-validation).

Lorsque `cv` n'est pas défini, `cross_val_score()` renverra un tableau de 5 nombres par défaut (`cv=None` est identique à `cv=5`).

Rappelez-vous, vous pouvez voir les paramètres d'une fonction en utilisant **SHIFT + TAB** (dans un Jupyter Notebook) à l'intérieur des parenthèses.

Mais attendez, vous vous demandez peut-être, qu'est-ce que la validation croisée ?

Une visualisation pourrait aider.

<img src='https://github.com/mrdbourke/zero-to-mastery-ml/blob/master/images/sklearn-cross-validation.png?raw=true' width=600/>

Nous avons déjà traité la Figure 1.0 en utilisant `score(X_test, y_test)`.

Mais en examinant cela plus en profondeur, si un modèle est entraîné en utilisant les données d'entraînement ou 80% des échantillons, cela signifie que 20% des échantillons ne sont pas utilisés pour que le modèle apprenne quoi que ce soit.

Cela signifie également que selon les 80% utilisés pour l'entraînement et les 20% utilisés pour évaluer le modèle, il peut obtenir un score qui ne reflète pas l'ensemble du jeu de données.

Par exemple, si beaucoup d'exemples faciles se trouvent dans les 80% de données d'entraînement, lorsqu'il s'agit de tester sur les 20%, votre modèle peut mal performer.

L'inverse est également vrai.

La Figure 2.0 montre la validation croisée à 5 plis (5-fold cross-validation), une méthode qui tente d'apporter une solution à :

1. Ne pas s'entraîner sur toutes les données (toujours garder les ensembles d'entraînement et de test séparés).
2. Éviter d'obtenir des scores chanceux sur des divisions uniques des données.

Au lieu de s'entraîner uniquement sur 1 division d'entraînement et d'évaluer sur 1 division de test, la validation croisée à 5 plis le fait 5 fois.

Sur une division différente à chaque fois, renvoyant un score pour chacune.

Pourquoi 5 plis ?

Le nom réel de cette configuration est la validation croisée à K plis. Où K est un nombre arbitraire. Nous avons utilisé 5 parce que cela semble visuellement agréable, et c'est la valeur par défaut dans [`sklearn.model_selection.cross_val_score`](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.cross_val_score.html).

La Figure 2.0 est ce qui se passe lorsque nous exécutons ce qui suit.

````python
# Validation croisée à 5 plis
cross_val_score(clf, X, y, cv=5) # cv est équivalent à K
````

Puisque nous avons défini `cv=5` (validation croisée à 5 plis), nous obtenons 5 scores différents au lieu de 1.

Prendre la moyenne de ce tableau nous donne une idée plus approfondie de la façon dont notre modèle se comporte en convertissant les 5 scores en un seul.

````python
np.random.seed(42)

# Score d'une seule division d'entraînement et de test
clf_single_score = clf.score(X_test, y_test)

# Prendre la moyenne de la validation croisée à 5 plis
clf_cross_val_score = np.mean(cross_val_score(clf, X, y, cv=5))

clf_single_score, clf_cross_val_score
````

Remarquez que la moyenne de `cross_val_score()` est légèrement inférieure à la valeur unique renvoyée par `score()`.

Dans ce cas, si on vous demandait de rapporter la précision de votre modèle, même si elle est plus basse, vous préféreriez la métrique validée de manière croisée à la métrique non validée de manière croisée.

### 4.2.1 Métriques d'évaluation des modèles de classification

#### Aire sous la courbe ROC (Receiver Operating Characteristic)
Si cela semble compliqué, c'est parce que lire le nom complet l'est.

On l'appelle généralement AUC pour Area Under Curve (Aire sous la courbe) et la courbe dont on parle est la courbe ROC (Receiver Operating Characteristic).

Donc si vous entendez quelqu'un parler d'AUC ou de ROC, ils parlent probablement de ce qui suit.

Les courbes ROC sont une comparaison du taux de vrais positifs (tpr) par rapport au taux de faux positifs (fpr).

Pour clarifier :
* Vrai positif = le modèle prédit 1 quand la vérité est 1
* Faux positif = le modèle prédit 1 quand la vérité est 0
* Vrai négatif = le modèle prédit 0 quand la vérité est 0
* Faux négatif = le modèle prédit 0 quand la vérité est 1

Maintenant que nous savons cela, voyons-en un. Scikit-Learn vous permet de calculer les informations nécessaires pour une courbe ROC en utilisant la fonction [`roc_curve`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.roc_curve.html#sklearn.metrics.roc_curve).

````python
from sklearn.metrics import roc_curve

# Faire des prédictions avec des probabilités
y_probs = clf.predict_proba(X_test)

# Garder uniquement les probabilités de la classe positive
y_probs = y_probs[:, 1]

# Calculer fpr, tpr et les seuils
fpr, tpr, thresholds = roc_curve(y_test, y_probs)

# Vérifier le taux de faux positifs
fpr
````

Les regarder seuls n'a pas beaucoup de sens. Il est beaucoup plus facile de voir leur valeur visuellement.

Créons une fonction d'aide pour créer une courbe ROC étant donné les taux de faux positifs (`fpr`) et les taux de vrais positifs (`tpr`).

> **Remarque :** À partir de Scikit-Learn 1.2+, il existe une fonctionnalité pour tracer une courbe ROC. Vous pouvez la trouver sous [`sklearn.metrics.RocCurveDisplay`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.RocCurveDisplay.html#sklearn-metrics-roccurvedisplay).

````python
import matplotlib.pyplot as plt

def plot_roc_curve(fpr, tpr):
    """
    Trace une courbe ROC étant donné le taux de faux positifs (fpr) et
    le taux de vrais positifs (tpr) d'un classificateur.
    """
    # Tracer la courbe ROC
    plt.plot(fpr, tpr, color='orange', label='ROC')
    # Tracer la ligne sans pouvoir prédictif (ligne de base)
    plt.plot([0, 1], [0, 1], color='darkblue', linestyle='--', label='Aléatoire')
    # Personnaliser le graphique
    plt.xlabel('Taux de faux positifs')
    plt.ylabel('Taux de vrais positifs')
    plt.title('Courbe ROC (Receiver Operating Characteristic)')
    plt.legend()
    plt.show()
    
plot_roc_curve(fpr, tpr)
````

En regardant le graphique pour la première fois, cela peut sembler un peu déroutant.

La principale chose à retenir ici est que notre modèle fait beaucoup mieux que de deviner.

Une métrique que vous pouvez utiliser pour quantifier la courbe ROC en un seul nombre est l'AUC (Area Under Curve).

Scikit-Learn implémente une fonction pour calculer cela appelée [`sklearn.metrics.roc_auc_score`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.roc_auc_score.html#sklearn.metrics.roc_auc_score).

Le score ROC AUC maximum que vous pouvez atteindre est 1,0 et généralement, plus il est proche de 1,0, meilleur est le modèle.

````python
from sklearn.metrics import roc_auc_score

roc_auc_score_value = roc_auc_score(y_test, y_probs)
roc_auc_score_value
````

Bien que ce soit un bon exercice, nous n'avions pas vraiment besoin de créer notre propre fonction `plot_roc_curve`.

Scikit-Learn nous permet de tracer une courbe ROC directement à partir de notre estimateur/modèle en utilisant la méthode de classe [`sklearn.metrics.RocCurveDisplay.from_estimator`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.RocCurveDisplay.html#sklearn.metrics.RocCurveDisplay.from_estimator) et en lui passant notre `estimator`, `X_test` et `y_test`.

#### Rapport de classification

La dernière métrique majeure que vous devriez considérer lors de l'évaluation d'un modèle de classification est un rapport de classification.

Un rapport de classification est davantage une collection de métriques qu'une seule métrique.

Vous pouvez créer un rapport de classification en utilisant la méthode [`sklearn.metrics.classification_report`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.classification_report.html) de Scikit-Learn.

Voyons-en un.

````python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_preds))
````

Il renvoie quatre colonnes : précision, rappel, score f1 et support.

Le nombre de lignes dépendra du nombre de classes différentes. Mais il y aura toujours trois lignes étiquetées accuracy, macro avg et weighted avg.

Chaque terme mesure quelque chose de légèrement différent :
* **Précision (Precision)** - Indique la proportion d'identifications positives (le modèle a prédit la classe `1`) qui étaient réellement correctes. Un modèle qui ne produit pas de faux positifs a une précision de 1,0.
* **Rappel (Recall)** - Indique la proportion de positifs réels qui ont été correctement classés. Un modèle qui ne produit pas de faux négatifs a un rappel de 1,0.
* **Score F1 (F1 score)** - Une combinaison de précision et de rappel. Un modèle parfait atteint un score F1 de 1,0.
* **Support (Support)** - Le nombre d'échantillons sur lesquels chaque métrique a été calculée.
* **Précision (Accuracy)** - La précision du modèle sous forme décimale. Une précision parfaite est égale à 1,0, en d'autres termes, obtenir la prédiction correcte 100% du temps.
* **Moyenne macro (Macro avg)** - Abréviation de moyenne macro, la moyenne de la précision, du rappel et du score F1 entre les classes. La moyenne macro ne prend pas en compte le déséquilibre des classes. Donc si vous avez des déséquilibres de classes (plus d'exemples d'une classe que d'une autre), vous devriez y prêter attention.
* **Moyenne pondérée (Weighted avg)** - Abréviation de moyenne pondérée, la moyenne pondérée de la précision, du rappel et du score F1 entre les classes. Pondéré signifie que chaque métrique est calculée en tenant compte du nombre d'échantillons dans chaque classe. Cette métrique favorisera la classe majoritaire (par exemple, elle donnera une valeur élevée lorsqu'une classe surpasse une autre en raison d'un plus grand nombre d'échantillons).

Quand devriez-vous utiliser chacune ?

Il peut être tentant de baser les performances de vos modèles de classification uniquement sur la précision. Et la précision est une bonne métrique à rapporter, sauf lorsque vous avez des classes très déséquilibrées.

Par exemple, disons qu'il y avait 10 000 personnes. Et 1 d'entre elles avait une maladie. On vous demande de construire un modèle pour prédire qui l'a.

Vous construisez le modèle et constatez que votre modèle est précis à 99,99%. Ce qui semble génial !
...jusqu'à ce que vous réalisiez que tout ce qu'il fait, c'est prédire que personne n'a la maladie, en d'autres termes, les 10 000 prédictions sont fausses.

Dans ce cas, vous voudriez vous tourner vers des métriques telles que la précision, le rappel et le score F1.

````python
# Où la précision et le rappel deviennent précieux
disease_true = np.zeros(10000)
disease_true[0] = 1 # un seul cas

disease_preds = np.zeros(10000) # chaque prédiction est 0

pd.DataFrame(classification_report(disease_true, 
                                   disease_preds, 
                                   output_dict=True,
                                   zero_division=0))
````

Vous pouvez voir ici que nous avons une précision de 0,9999 (99,99%), une excellente précision et un excellent rappel pour la classe 0.0, mais rien pour la classe 1.0.

Demandez-vous, bien que le modèle atteigne une précision de 99,99%, est-il utile ?

Pour résumer :
* La précision est une bonne mesure pour commencer si toutes les classes sont équilibrées (par exemple, même quantité d'échantillons étiquetés avec 0 ou 1)
* La précision et le rappel deviennent plus importants lorsque les classes sont déséquilibrées.
* Si les prédictions de faux positifs sont pires que les faux négatifs, visez une précision plus élevée.
* Si les prédictions de faux négatifs sont pires que les faux positifs, visez un rappel plus élevé.

> **Ressource :** Pour en savoir plus sur la précision et le rappel et les compromis entre eux, je vous suggère de consulter le [guide Precision-Recall de Scikit-Learn](https://scikit-learn.org/stable/auto_examples/model_selection/plot_precision_recall.html).

### 4.2.2 Métriques d'évaluation des modèles de régression

Comme pour la classification, il existe [plusieurs métriques que vous pouvez utiliser pour évaluer vos modèles de régression](https://scikit-learn.org/stable/modules/model_evaluation.html#regression-metrics).

Nous examinerons les suivantes.

1. **R² (prononcé r-carré) ou coefficient de détermination** - Compare les prédictions de votre modèle à la moyenne des cibles. Les valeurs peuvent aller de moins l'infini (un très mauvais modèle) à 1. Par exemple, si tout ce que fait votre modèle est de prédire la moyenne des cibles, sa valeur R² serait de 0. Et si votre modèle prédit parfaitement une série de nombres, sa valeur R² serait de 1. Plus c'est élevé, mieux c'est.
2. **Erreur absolue moyenne (MAE)** - La moyenne des différences absolues entre les prédictions et les valeurs réelles. Cela vous donne une idée de l'ampleur des erreurs de vos prédictions. Plus c'est bas, mieux c'est.
3. **Erreur quadratique moyenne (MSE)** - La moyenne des différences au carré entre les prédictions et les valeurs réelles. Élever les erreurs au carré élimine les erreurs négatives. Cela amplifie également les valeurs aberrantes (échantillons qui ont des erreurs plus importantes). Plus c'est bas, mieux c'est.

Voyons-les en action. Tout d'abord, nous allons reprendre notre code de modèle de régression.

````python
# Importer la classe de modèle RandomForestRegressor du module ensemble
from sklearn.ensemble import RandomForestRegressor

# Configurer la graine aléatoire
np.random.seed(42)

# Diviser les données en caractéristiques (X) et étiquettes (y)
X = housing_df.drop("target", axis=1)
y = housing_df["target"]

# Diviser en ensembles d'entraînement et de test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Instancier et ajuster le modèle (sur l'ensemble d'entraînement)
model = RandomForestRegressor()
model.fit(X_train, y_train);
````

**Score R² (coefficient de détermination)**

Une fois que vous avez un modèle de régression entraîné, la métrique d'évaluation par défaut dans la fonction `score()` est R².

````python
# Calculer le score R² du modèle
model.score(X_test, y_test)
````

En dehors de la fonction `score()`, R² peut être calculé à l'aide de la fonction [`r2_score()`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.r2_score.html#sklearn.metrics.r2_score) de Scikit-Learn.

Un modèle qui ne prédirait que la moyenne obtiendrait un score de 0.

````python
from sklearn.metrics import r2_score

# Remplir un tableau avec la moyenne de y_test
y_test_mean = np.full(len(y_test), y_test.mean())

r2_score(y_test, y_test_mean)
````

Et un modèle parfait obtiendrait un score de 1.

````python
r2_score(y_test, y_test)
````

Pour vos modèles de régression, vous voudrez maximiser R², tout en minimisant MAE et MSE.

**Erreur absolue moyenne (MAE)**

L'erreur absolue moyenne d'un modèle peut être calculée avec la méthode [`sklearn.metrics.mean_absolute_error`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.mean_absolute_error.html) de Scikit-Learn.

````python
# Erreur absolue moyenne
from sklearn.metrics import mean_absolute_error

y_preds = model.predict(X_test)
mae = mean_absolute_error(y_test, y_preds)
mae
````

Notre modèle atteint un MAE de 0,327.

Cela signifie qu'en moyenne, les prédictions de notre modèle sont à 0,327 unités de la valeur réelle.

Rendons cela un peu plus visuel.

````python
df = pd.DataFrame(data={"valeurs réelles": y_test, 
                   "prédictions": y_preds})

df
````

Vous pouvez voir que les prédictions sont légèrement différentes des valeurs réelles.

Selon le problème sur lequel vous travaillez, avoir une différence comme celle que nous avons maintenant pourrait être acceptable. D'un autre côté, cela pourrait aussi ne pas être acceptable, ce qui signifie que les prédictions devraient être plus proches.

````python
fig, ax = plt.subplots()
x = np.arange(0, len(df), 1)
ax.scatter(x, df["valeurs réelles"], c='b', label="Valeurs réelles")
ax.scatter(x, df["prédictions"], c='r', label="Prédictions")
ax.legend(loc=(1, 0.5));
````

**Erreur quadratique moyenne (MSE)**

Qu'en est-il de MSE ?

Nous pouvons le calculer avec [`sklearn.metrics.mean_squared_error`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.mean_squared_error.html) de Scikit-Learn.

````python
# Erreur quadratique moyenne
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(y_test, y_preds)
mse
````

MSE sera souvent plus élevé que MAE car il élève les erreurs au carré plutôt que de ne prendre en compte que la différence absolue.

Maintenant, vous vous demandez peut-être quelle métrique d'évaluation de régression vous devriez utiliser ?

* R² est similaire à la précision. Il vous donne une indication rapide de la performance potentielle de votre modèle. Généralement, plus votre valeur R² est proche de 1,0, meilleur est le modèle. Mais cela ne vous dit pas exactement à quel point votre modèle est erroné en termes d'écart entre chaque prédiction.
* MAE donne une meilleure indication de l'écart moyen entre chacune des prédictions de votre modèle.
* Quant à MAE ou MSE, en raison de la façon dont MSE est calculé, en élevant au carré les différences entre les valeurs prédites et les valeurs réelles, il amplifie les différences plus importantes. Disons que nous prédisons la valeur des maisons (ce que nous faisons).
    * Accordez plus d'attention à MAE : Lorsque être à 10 000 $ d'écart est ***deux fois*** pire qu'être à 5 000 $ d'écart.
    * Accordez plus d'attention à MSE : Lorsque être à 10 000 $ d'écart est ***plus de deux fois*** pire qu'être à 5 000 $ d'écart.
    
> **Remarque :** Ce que nous avons couvert ici n'est qu'une poignée de métriques potentielles que vous pouvez utiliser pour évaluer vos modèles. Si vous recherchez une liste complète, consultez la [documentation sur les métriques et le scoring de Scikit-Learn](https://scikit-learn.org/stable/modules/model_evaluation.html).

### 4.2.3 Évaluer un modèle à l'aide du paramètre `scoring`

Nous avons couvert de nombreuses façons d'évaluer les prédictions d'un modèle, mais nous n'avons même pas abordé le paramètre `scoring`...

Ne vous inquiétez pas, c'est très similaire à ce que nous avons fait !

Pour rappel, le paramètre `scoring` peut être utilisé avec une fonction comme `cross_val_score()` pour indiquer à Scikit-Learn quelle métrique d'évaluation renvoyer en utilisant la validation croisée.

Examinons-le avec notre modèle de classification et le jeu de données sur les maladies cardiaques.

````python
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import RandomForestClassifier

np.random.seed(42)

X = heart_disease.drop("target", axis=1)
y = heart_disease["target"]

clf = RandomForestClassifier(n_estimators=100)
````

D'abord, nous utiliserons la valeur par défaut, qui est la précision moyenne.

````python
np.random.seed(42)
cv_acc = cross_val_score(clf, X, y, cv=5)
cv_acc
````

Nous avons déjà vu cela, maintenant nous avons 5 scores de précision différents sur différentes divisions de test des données.

La moyenne de ces scores donne la précision validée de manière croisée.

````python
# Précision validée de manière croisée
print(f"La précision validée de manière croisée est : {np.mean(cv_acc)*100:.2f}%")
````

Nous pouvons trouver la même chose en utilisant le paramètre `scoring` et en lui passant `"accuracy"`.

````python
np.random.seed(42)
cv_acc = cross_val_score(clf, X, y, cv=5, scoring="accuracy")
print(f"La précision validée de manière croisée est : {np.mean(cv_acc)*100:.2f}%")
````

Il en va de même pour les autres métriques que nous avons utilisées pour la classification.

Essayons `"precision"`.

````python
np.random.seed(42)
cv_precision = cross_val_score(clf, X, y, cv=5, scoring="precision")
print(f"La précision (precision) validée de manière croisée est : {np.mean(cv_precision):.2f}")
````

Qu'en est-il de `"recall"` ?

````python
np.random.seed(42)
cv_recall = cross_val_score(clf, X, y, cv=5, scoring="recall")
print(f"Le rappel (recall) validé de manière croisée est : {np.mean(cv_recall):.2f}")
````

Et `"f1"` (pour le score F1) ?

````python
np.random.seed(42)
cv_f1 = cross_val_score(clf, X, y, cv=5, scoring="f1")
print(f"Le score F1 validé de manière croisée est : {np.mean(cv_f1):.2f}")
````

Nous pouvons répéter ce processus avec nos métriques de régression.

Revenons à notre modèle de régression.

````python
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import RandomForestRegressor

np.random.seed(42)

X = housing_df.drop("target", axis=1)
y = housing_df["target"]

model = RandomForestRegressor(n_estimators=100)
````

La valeur par défaut est `"r2"`.

> **Remarque :** Nous pouvons chronométrer le temps d'exécution d'une seule cellule de code en utilisant la [commande magique `%%time`](https://ipython.readthedocs.io/en/stable/interactive/magics.html#magic-time).

````python
%%time 
np.random.seed(42)
cv_r2 = cross_val_score(model, X, y, cv=5, scoring="r2")
print(f"Le score R^2 validé de manière croisée est : {np.mean(cv_r2):.2f}")
````

Mais nous pouvons utiliser `"neg_mean_absolute_error"` pour MAE (erreur absolue moyenne).

````python
%%time
np.random.seed(42)
cv_mae = cross_val_score(model, X, y, cv=5, scoring="neg_mean_absolute_error")
print(f"Le score MAE validé de manière croisée est : {np.mean(cv_mae):.2f}")
````

Pourquoi le `"neg_"` ?

Parce que la documentation de Scikit-Learn indique :
> ["Tous les objets de scoring suivent la convention selon laquelle des valeurs de retour plus élevées sont meilleures que des valeurs de retour plus basses."](https://scikit-learn.org/stable/modules/model_evaluation.html#common-cases-predefined-values)

Ce qui, dans ce cas, signifie qu'une valeur négative plus basse (plus proche de 0) est meilleure.

Qu'en est-il de `"neg_mean_squared_error"` pour MSE (erreur quadratique moyenne) ?

````python
np.random.seed(42)
cv_mse = cross_val_score(model, 
                         X, 
                         y, 
                         cv=5,
                         scoring="neg_mean_squared_error")
print(f"Le score MSE validé de manière croisée est : {np.mean(cv_mse):.2f}")
````

### 4.3 Utilisation de différentes métriques d'évaluation avec Scikit-Learn

Vous vous souvenez de la troisième façon d'évaluer les fonctions Scikit-Learn ?

> 3. Fonctions de métrique spécifiques au problème. Similaire à la façon dont le paramètre `scoring` peut recevoir différentes fonctions de scoring, Scikit-Learn implémente celles-ci en tant que fonctions autonomes.

Eh bien, nous avons en quelque sorte couvert cette troisième façon d'utiliser les métriques d'évaluation avec Scikit-Learn.

#### Fonctions de classification
Pour :
* La précision (accuracy), nous pouvons utiliser [`sklearn.metrics.accuracy_score`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.accuracy_score.html)
* La précision (precision), nous pouvons utiliser [`sklearn.metrics.precision_score`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.precision_score.html)
* Le rappel (recall), nous pouvons utiliser [`sklearn.metrics.recall_score`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.recall_score.html)
* F1, nous pouvons utiliser [`sklearn.metrics.f1_score`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.f1_score.html)

````python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split

np.random.seed(42)

X = heart_disease.drop("target", axis=1)
y = heart_disease["target"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

clf = RandomForestClassifier(n_estimators=100)
clf.fit(X_train, y_train)

# Faire des prédictions
y_preds = clf.predict(X_test)

# Évaluer le classificateur
print("Métriques du classificateur sur l'ensemble de test :")
print(f"Précision (accuracy) : {accuracy_score(y_test, y_preds) * 100:.2f}%")
print(f"Précision (precision) : {precision_score(y_test, y_preds):.2f}")
print(f"Rappel (recall) : {recall_score(y_test, y_preds):.2f}")
print(f"F1 : {f1_score(y_test, y_preds):.2f}")
````

#### Métriques de régression

Nous pouvons utiliser une configuration similaire pour notre problème de régression, mais avec des méthodes différentes.

Pour :
* R², nous pouvons utiliser [`sklearn.metrics.r2_score`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.r2_score.html)
* MAE (erreur absolue moyenne), nous pouvons utiliser [`sklearn.metrics.mean_absolute_error`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.mean_absolute_error.html)
* MSE (erreur quadratique moyenne), nous pouvons utiliser [`sklearn.metrics.mean_squared_error`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.mean_squared_error.html)

````python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split

np.random.seed(42)

X = housing_df.drop("target", axis=1)
y = housing_df["target"]

X_train, X_test, y_train, y_test = train_test_split(X, 
                                                    y, 
                                                    test_size=0.2)

model = RandomForestRegressor(n_estimators=100, 
                              n_jobs=-1)
model.fit(X_train, y_train)

# Faire des prédictions
y_preds = model.predict(X_test)

# Évaluer le modèle
print("Métriques du modèle de régression sur l'ensemble de test :")
print(f"R^2 : {r2_score(y_test, y_preds):.2f}")
print(f"MAE : {mean_absolute_error(y_test, y_preds):.2f}")
print(f"MSE : {mean_squared_error(y_test, y_preds):.2f}")
````

### 5.1 Réglage des hyperparamètres à la main

Jusqu'à présent, nous avons travaillé avec des jeux de données d'entraînement et de test.

Vous entraînez un modèle sur un ensemble d'entraînement et l'évaluez sur un jeu de données de test.

Mais le réglage des hyperparamètres introduit un troisième ensemble, un ensemble de validation.

Maintenant, le processus devient :
1. Entraîner un modèle sur les données d'entraînement.
2. (Essayer d') améliorer les hyperparamètres du modèle sur l'ensemble de validation.
3. Évaluer le modèle sur l'ensemble de test.

Si notre jeu de données de départ contenait 100 dossiers de patients différents indiquant qui avait une maladie cardiaque et qui n'en avait pas, et que nous voulions construire un modèle d'apprentissage automatique pour prédire qui avait une maladie cardiaque et qui n'en avait pas, cela pourrait ressembler à ceci :

<img src="https://github.com/mrdbourke/zero-to-mastery-ml/blob/master/images/sklearn-train-valid-test-annotated.png?raw=true" width=500/>

Puisque nous savons que nous utilisons un `RandomForestClassifier` et que nous connaissons les hyperparamètres que nous voulons ajuster, voyons à quoi cela ressemble.

Tout d'abord, rappelons-nous les paramètres de base.

````python
clf.get_params()
````

Et nous allons ajuster :
* `max_depth`
* `max_features`
* `min_samples_leaf`
* `min_samples_split`
* `n_estimators`

Nous utiliserons le même code qu'auparavant, sauf que cette fois, nous créerons une division en ensembles d'entraînement, de validation et de test.

L'ensemble d'entraînement contiendra 70% des données et les ensembles de validation et de test contiendront chacun 15%.

Obtenons quelques résultats de référence, puis nous réglerons le modèle.

Et puisque nous allons évaluer quelques modèles, créons une fonction d'évaluation.

````python
def evaluate_preds(y_true: np.array, 
                   y_preds: np.array) -> dict:
    """
    Effectue une comparaison d'évaluation entre les étiquettes y_true et les étiquettes y_pred.

    Renvoie plusieurs métriques sous forme de dictionnaire.
    """
    accuracy = accuracy_score(y_true, y_preds)
    precision = precision_score(y_true, y_preds)
    recall = recall_score(y_true, y_preds)
    f1 = f1_score(y_true, y_preds)
    metric_dict = {"accuracy": round(accuracy, 2),
                   "precision": round(precision, 2), 
                   "recall": round(recall, 2),
                   "f1": round(f1, 2)}
    print(f"Précision : {accuracy * 100:.2f}%")
    print(f"Précision (precision) : {precision:.2f}")
    print(f"Rappel (recall) : {recall:.2f}")
    print(f"Score F1 : {f1:.2f}")

    return metric_dict
````

### 5.2 Réglage des hyperparamètres avec [`RandomizedSearchCV`](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.RandomizedSearchCV.html)

Le [`sklearn.model_selection.RandomizedSearchCV`](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.RandomizedSearchCV.html) de Scikit-Learn nous permet de rechercher aléatoirement parmi différents hyperparamètres pour voir lesquels fonctionnent le mieux.

Il stocke également des détails sur ceux qui fonctionnent le mieux !

Voyons-le en action.

Tout d'abord, nous créons un dictionnaire de distributions de paramètres (collections de différentes valeurs pour des hyperparamètres spécifiques) que nous aimerions explorer.

Ce dictionnaire se présente sous la forme :

```python
param_distributions = {"nom_hyperparamètre": [valeurs_à_essayer_aléatoirement]}
```

Où `"nom_hyperparamètre"` est la valeur d'un hyperparamètre spécifique pour un modèle et `[valeurs_à_essayer_aléatoirement]` est une liste de valeurs pour cet hyperparamètre spécifique à essayer aléatoirement.

````python
# Grille d'hyperparamètres que RandomizedSearchCV va explorer
param_distributions = {"n_estimators": [10, 100, 200, 500, 1000, 1200],
                       "max_depth": [None, 5, 10, 20, 30],
                       "max_features": ["sqrt", "log2", None],
                       "min_samples_split": [2, 4, 6, 8],
                       "min_samples_leaf": [1, 2, 4, 8]}
````

D'où viennent ces valeurs ?

Elles sont inventées.

Inventées ?

Oui.

Pas complètement sorties de nulle part, mais après avoir lu la [documentation Scikit-Learn sur les Forêts Aléatoires](https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.RandomForestClassifier.html), vous verrez que certaines de ces valeurs ont des valeurs qui fonctionnent généralement bien et certains hyperparamètres prennent des chaînes de caractères plutôt que des entiers.

Maintenant que nous avons configuré le dictionnaire de distribution des paramètres, le `RandomizedSearchCV` de Scikit-Learn va l'examiner, choisir une valeur aléatoire pour chacun, instancier un modèle avec ces valeurs et tester chaque modèle.

Combien de modèles va-t-il tester ?

Autant qu'il y a de combinaisons d'hyperparamètres à tester. Additionnons-les.

````python
# Compter le nombre total de combinaisons d'hyperparamètres à tester
total_randomized_hyperparameter_combintions_to_test = np.prod([len(value) for value in param_distributions.values()])
print(f"Il y a {total_randomized_hyperparameter_combintions_to_test} combinaisons potentielles d'hyperparamètres à tester.")
````

Woah !

C'est beaucoup de combinaisons !

Ou...

Nous pouvons définir le paramètre `n_iter` pour limiter le nombre de modèles que `RandomizedSearchCV` teste (par exemple, `n_iter=20` signifie essayer `20` combinaisons aléatoires différentes d'hyperparamètres et validera chaque ensemble de manière croisée, donc si `cv=5`, 5x20 = 100 entraînements au total).

Le meilleur ?

Les résultats que nous obtenons seront validés de manière croisée (d'où le CV dans `RandomizedSearchCV`), nous pouvons donc utiliser `train_test_split()`.

Et puisque nous passons en revue tant de modèles différents, nous définirons `n_jobs=-1` dans notre [`RandomForestClassifier`](https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.RandomForestClassifier.html) pour que Scikit-Learn tire parti de tous les cœurs (processeurs) de nos ordinateurs.

Voyons cela en action.

> **Remarque :** Selon `n_iter` (combien de modèles vous testez), les différentes valeurs dans la grille d'hyperparamètres et la puissance de votre ordinateur, l'exécution de la cellule ci-dessous peut prendre un certain temps (à titre de référence, cela a pris environ ~1 minute sur mon MacBook Pro M1 Pro).

````python
# Démarrer le chronomètre
import time
start_time = time.time()

from sklearn.model_selection import RandomizedSearchCV, train_test_split

np.random.seed(42)

# Diviser en X et y
X = heart_disease.drop("target", axis=1)
y = heart_disease["target"]

# Diviser en ensembles d'entraînement et de test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Définir n_jobs à -1 pour utiliser tous les cœurs disponibles sur votre machine (si cela cause des erreurs, essayez n_jobs=1)
clf = RandomForestClassifier(n_jobs=-1)

# Configurer RandomizedSearchCV
n_iter = 10 # essayer 10 modèles au total
rs_clf = RandomizedSearchCV(estimator=clf,
                            param_distributions=param_distributions,
                            n_iter=n_iter, 
                            cv=5, # validation croisée à 5 plis
                            verbose=2) # afficher les résultats

# Ajuster la version RandomizedSearchCV de clf (fait la validation croisée pour nous, donc pas besoin d'utiliser un ensemble de validation)
rs_clf.fit(X_train, y_train);

# Terminer le chronomètre
end_time = time.time()
print(f"[INFO] Temps total pris pour {n_iter} combinaisons aléatoires d'hyperparamètres : {end_time - start_time:.2f} secondes.")
````

Lorsque `RandomizedSearchCV` parcourt les `n_iter` combinaisons de l'espace de recherche des hyperparamètres, il stocke les meilleurs dans l'attribut `best_params_`.

````python
# Trouver les meilleurs hyperparamètres trouvés par RandomizedSearchCV
rs_clf.best_params_
````

Maintenant, lorsque nous appelons `predict()` sur `rs_clf` (notre version `RandomizedSearchCV` de notre classificateur), il utilisera les meilleurs hyperparamètres qu'il a trouvés.

````python
# Faire des prédictions avec les meilleurs hyperparamètres
rs_y_preds = rs_clf.predict(X_test)

# Évaluer les prédictions
rs_metrics = evaluate_preds(y_test, rs_y_preds)
````

Excellent ! 

Grâce à `RandomizedSearchCV` qui a testé un grand nombre de différents hyperparamètres, nous obtenons une belle amélioration de toutes les métriques d'évaluation pour notre modèle de classification.

### 5.3 Réglage des hyperparamètres avec [`GridSearchCV`](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.GridSearchCV.html)

Il existe une autre façon d'essayer d'améliorer les hyperparamètres de notre modèle.

C'est avec [`sklearn.model_selection.GridSearchCV`](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.GridSearchCV.html).

La principale différence entre `GridSearchCV` et `RandomizedSearchCV` est que `GridSearchCV` recherche de manière exhaustive dans une grille d'hyperparamètres (il essaiera toutes les combinaisons possibles), tandis que `RandomizedSearchCV` recherche aléatoirement dans une grille d'hyperparamètres (s'arrêtant après `n_iter` combinaisons).

`GridSearchCV` fait également référence à un dictionnaire de distributions de paramètres comme une grille de paramètres (via le paramètre `param_grid`).

Par exemple, examinons notre dictionnaire d'hyperparamètres.

````python
param_distributions
````

`RandomizedSearchCV` essaie `n_iter` combinaisons de différentes valeurs.

Alors que `GridSearchCV` essaiera toutes les combinaisons possibles.

Et si vous vous souvenez du calcul précédent : `max_depth` a 4 valeurs, `max_features` en a 2, `min_samples_leaf` en a 3, `min_samples_split` en a 3, `n_estimators` en a 5.

Cela fait 4x2x3x3x5 = 360 modèles !

Cela pourrait prendre beaucoup de temps selon la puissance de l'ordinateur que vous utilisez, la quantité de données dont vous disposez et la complexité des hyperparamètres (généralement des valeurs plus élevées signifient un modèle plus complexe).

Dans notre cas, les données que nous utilisons sont relativement petites (seulement ~300 échantillons).

Puisque nous avons déjà essayé de trouver des hyperparamètres idéaux en utilisant `RandomizedSearchCV`, nous allons créer une autre grille d'hyperparamètres basée sur les `best_params_` de `rs_clf` avec moins d'options, puis essayer d'utiliser `GridSearchCV` pour trouver un ensemble plus idéal.

En essence, le flux de travail pourrait être :
1. Régler les hyperparamètres à la main pour avoir une idée des données/du modèle.
2. Créer un grand ensemble de distributions d'hyperparamètres et les rechercher aléatoirement avec `RandomizedSearchCV`.
3. Trouver les meilleurs hyperparamètres de l'étape 2 et réduire l'espace de recherche avant de rechercher de manière exhaustive dans un sous-ensemble plus petit avec `GridSearchCV`.

> **Remarque :** Basé sur les `best_params_` de `rs_clf`, cela implique que le prochain ensemble d'hyperparamètres que nous essaierons est à peu près dans la même plage que le meilleur ensemble trouvé par `RandomizedSearchCV`.

````python
# Créer une grille d'hyperparamètres similaire à rs_clf.best_params_
param_grid = {"n_estimators": [200, 1000],
              "max_depth": [30, 40, 50],
              "max_features": ["log2"],
              "min_samples_split": [2, 4, 6, 8],
              "min_samples_leaf": [4]}
````

Nous avons créé une autre grille d'hyperparamètres à explorer, cette fois avec moins d'options au total.

````python
# Compter le nombre total de combinaisons d'hyperparamètres à tester
total_grid_search_hyperparameter_combinations_to_test = np.prod([len(value) for value in param_grid.values()])
print(f"Il y a {total_grid_search_hyperparameter_combinations_to_test} combinaisons d'hyperparamètres à tester.")
print(f"C'est {total_randomized_hyperparameter_combintions_to_test/total_grid_search_hyperparameter_combinations_to_test} fois moins\
 qu'avant (précédemment : {total_randomized_hyperparameter_combintions_to_test}).")
````

Maintenant, lorsque nous exécutons `GridSearchCV`, en lui passant notre classificateur (`clf`), la grille de paramètres (`param_grid`) et le nombre de plis de validation croisée que nous souhaitons utiliser (`cv=5`), il créera un modèle avec chaque combinaison d'hyperparamètres, puis validera chacun 5 fois (par exemple, 36 combinaisons d'hyperparamètres * 5 = 135 entraînements au total) et vérifiera les résultats.

> **Remarque :** Selon la puissance de calcul de la machine que vous utilisez, la cellule suivante peut prendre quelques minutes à s'exécuter (à titre de référence, cela a pris environ 60 secondes sur mon MacBook Pro M1 Pro).

````python
# Démarrer le chronomètre
import time
start_time = time.time()

from sklearn.model_selection import GridSearchCV, train_test_split

np.random.seed(42)

# Diviser en X & y
X = heart_disease.drop("target", axis=1)
y = heart_disease["target"]

# Diviser en ensembles d'entraînement et de test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Définir n_jobs à -1 pour utiliser tous les cœurs disponibles de la machine (si cela produit des erreurs, essayez n_jobs=1)
clf = RandomForestClassifier(n_jobs=-1)

# Configurer GridSearchCV
gs_clf = GridSearchCV(estimator=clf,
                      param_grid=param_grid,
                      cv=5, # validation croisée à 5 plis
                      verbose=2) # afficher la progression

# Ajuster la version RandomizedSearchCV de clf
gs_clf.fit(X_train, y_train);

# Trouver le temps d'exécution
end_time = time.time()
````

````python
# Combien de temps cela a-t-il pris ? 
total_time = end_time - start_time
print(f"[INFO] Le temps d'exécution total pour GridSearchCV était de {total_time:.2f} secondes.")
````

Une fois terminé, nous pouvons vérifier les meilleures combinaisons d'hyperparamètres trouvées en utilisant l'attribut `best_params_`.

````python
# Vérifier les meilleurs hyperparamètres trouvés avec GridSearchCV
gs_clf.best_params_
````

Et par défaut, lorsque nous appelons la fonction `predict()` sur `gs_clf`, elle utilisera les meilleurs hyperparamètres.

````python
# Faire des prédictions avec le classificateur GridSearchCV
gs_y_preds = gs_clf.predict(X_test)

# Évaluer les prédictions
gs_metrics = evaluate_preds(y_test, gs_y_preds)
gs_metrics
````

## 6. Sauvegarde et chargement de modèles d'apprentissage automatique entraînés

Notre modèle `GridSearchCV` (`gs_clf`) a les meilleurs résultats jusqu'à présent, nous allons l'exporter et le sauvegarder dans un fichier.

### 6.1 Sauvegarde et chargement d'un modèle avec `pickle`

Nous avons vu tout au début qu'une façon de sauvegarder un modèle est d'utiliser le [module `pickle` de Python](https://docs.python.org/3/library/pickle.html).

Nous utiliserons la méthode `dump()` de `pickle` et lui passerons notre modèle, `gs_clf`, ainsi que la fonction `open()` contenant une chaîne pour le nom de fichier sous lequel nous voulons sauvegarder notre modèle, ainsi que la chaîne `"wb"` qui signifie "write binary" (écrire en binaire), qui est le type de fichier sous lequel `open()` écrira notre modèle.

````python
import pickle

# Sauvegarder un modèle existant dans un fichier
best_model_file_name_pickle = "gs_random_forest_model_1.pkl" # l'extension .pkl signifie "pickle"
pickle.dump(gs_clf, open(best_model_file_name_pickle, "wb"))
````

Une fois sauvegardé, nous pouvons l'importer en utilisant la fonction `load()` de `pickle`, en lui passant `open()` contenant le nom du fichier sous forme de chaîne et `"rb"` signifiant "read binary" (lire en binaire).

````python
# Charger un modèle sauvegardé
loaded_pickle_model = pickle.load(open(best_model_file_name_pickle, "rb"))
````

Une fois que vous avez réimporté votre modèle entraîné en utilisant `pickle`, vous pouvez l'utiliser pour faire des prédictions comme d'habitude.

````python
# Faire des prédictions et évaluer le modèle chargé
pickle_y_preds = loaded_pickle_model.predict(X_test)
loaded_pickle_model_metrics = evaluate_preds(y_test, pickle_y_preds)
loaded_pickle_model_metrics
````

Vous remarquerez que les métriques d'évaluation du modèle réimporté sont les mêmes que celles du modèle avant que nous l'exportions.

````python
loaded_pickle_model_metrics == gs_metrics
````

Pour sauvegarder un modèle, nous pouvons utiliser la fonction `dump()` de `joblib`, en lui passant le modèle (`gs_clf`) et le nom de fichier souhaité.

````python
from joblib import dump, load

# Sauvegarder un modèle dans un fichier
best_model_file_name_joblib = "gs_random_forest_model_1.joblib"
dump(gs_clf, filename=best_model_file_name_joblib) 
````

````python
# Importer un modèle joblib sauvegardé
loaded_joblib_model = load(filename=best_model_file_name_joblib)
````

Encore une fois, une fois importé, nous pouvons faire des prédictions avec notre modèle.

````python
# Faire et évaluer les prédictions joblib 
joblib_y_preds = loaded_joblib_model.predict(X_test)
loaded_joblib_model_metrics = evaluate_preds(y_test, joblib_y_preds)
loaded_joblib_model_metrics
````

Et encore une fois, vous remarquerez que les métriques d'évaluation sont les mêmes qu'avant.

````python
loaded_joblib_model_metrics == gs_metrics
````

Alors lequel devriez-vous utiliser, `pickle` ou `joblib` ?

Selon la [documentation sur la persistance des modèles de Scikit-Learn](https://scikit-learn.org/stable/model_persistence.html), ils suggèrent qu'il peut être plus efficace d'utiliser `joblib` car il est plus efficace avec les grands tableaux numpy (qui peuvent être contenus dans les modèles Scikit-Learn entraînés/ajustés).

### 7.1 Création d'un [`Pipeline`](https://scikit-learn.org/stable/modules/generated/sklearn.pipeline.Pipeline.html) de régression
Vous vous souvenez peut-être que, dans la section 2 : Préparation des données, lorsque nous avons traité les données de ventes de voitures, pour construire un modèle de régression dessus, nous avons dû encoder les caractéristiques catégorielles en nombres et remplir les données manquantes.

La bonne nouvelle est que `Pipeline` peut nous aider à le nettoyer.

Rappelons-nous à quoi ressemble notre [`car-sales-extended-missing-data.csv`](https://github.com/mrdbourke/zero-to-mastery-ml/blob/master/data/car-sales-extended-missing-data.csv) sous forme de DataFrame.

Il y a 1000 lignes, trois caractéristiques sont catégorielles (`Make`, `Colour`, `Doors`), les deux autres sont numériques (`Odometer (KM)`, `Price`) et il y a 249 valeurs manquantes.

Nous allons devoir transformer les caractéristiques catégorielles en nombres et remplir les valeurs manquantes avant de pouvoir ajuster un modèle.

Nous allons construire un `Pipeline` pour le faire.

Le principal paramètre d'entrée de `Pipeline` est `steps`, qui est une liste de tuples (`[(nom_étape, action_à_prendre)]`) du nom de l'étape, plus l'action que vous souhaitez qu'il effectue.

Dans notre cas, vous pourriez considérer les étapes comme :
1. Remplir les données manquantes
2. Convertir les données en nombres
3. Construire un modèle sur les données

Allons-y !

````python
# Préparation des données
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

# Modélisation
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split, GridSearchCV

# Configuration de la graine aléatoire
import numpy as np
np.random.seed(42)

# Importer les données et supprimer les lignes avec des étiquettes manquantes
data = pd.read_csv("https://raw.githubusercontent.com/mrdbourke/zero-to-mastery-ml/master/data/car-sales-extended-missing-data.csv")
data.dropna(subset=["Price"], inplace=True)

# Définir différentes caractéristiques et pipelines de transformation
categorical_features = ["Make", "Colour"]
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="constant", fill_value="missing")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))])

door_feature = ["Doors"]
door_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="constant", fill_value=4))])

numeric_features = ["Odometer (KM)"]
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="mean"))
])

# Configurer les étapes de prétraitement (remplir les valeurs manquantes, puis convertir en nombres)
preprocessor = ColumnTransformer(
    transformers=[
        ("cat", categorical_transformer, categorical_features),
        ("door", door_transformer, door_feature),
        ("num", numeric_transformer, numeric_features)])

# Créer un pipeline de prétraitement et de modélisation
model = Pipeline(steps=[("preprocessor", preprocessor),
                        ("model", RandomForestRegressor(n_jobs=-1))])

# Diviser les données
X = data.drop("Price", axis=1)
y = data["Price"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Ajuster et évaluer le modèle
model.fit(X_train, y_train)
model.score(X_test, y_test)
````

Ce que nous avons fait, c'est combiner une série d'étapes de prétraitement des données (remplissage des valeurs manquantes, encodage des valeurs numériques) ainsi qu'un modèle dans un `Pipeline`.

Cela permet non seulement de nettoyer le code, mais aussi de s'assurer que les mêmes étapes sont effectuées à chaque exécution du code, plutôt que d'avoir plusieurs étapes de traitement différentes se déroulant à différentes étapes.

Il est également possible d'utiliser `GridSearchCV` ou `RandomizedSearchCV` avec un `Pipeline`.

La principale différence est que lors de la création d'une grille d'hyperparamètres, vous devez ajouter un préfixe à chaque hyperparamètre (voir la [documentation de `RandomForestRegressor`](https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.RandomForestRegressor.html) pour une liste complète des hyperparamètres possibles à régler).

Le préfixe est le nom de l'étape `Pipeline` que vous souhaitez modifier, suivi de deux traits de soulignement.

Par exemple, pour ajuster `n_estimators` de `"model"` dans le `Pipeline`, vous utiliseriez : `"model__n_estimators"` (notez le double trait de soulignement après `model__` au début).

Voyons cela !

> **Remarque :** Selon la puissance de traitement de votre ordinateur, la cellule ci-dessous peut prendre quelques minutes à s'exécuter. À titre de référence, cela a pris environ 60 secondes sur mon MacBook Pro M1 Pro.

````python
%%time

# Utilisation de la recherche par grille avec pipeline
pipe_grid = {
    "preprocessor__num__imputer__strategy": ["mean", "median"], # notez le double trait de soulignement après chaque préfixe "preprocessor__"
    "model__n_estimators": [100, 1000],
    "model__max_depth": [None, 5],
    "model__max_features": ["sqrt"],
    "model__min_samples_split": [2, 4]
}

gs_model = GridSearchCV(model, pipe_grid, cv=5, verbose=2)
gs_model.fit(X_train, y_train)
````

Maintenant, trouvons le score de notre modèle (par défaut, `GridSearchCV` sauvegarde le meilleur modèle dans l'objet `gs_model`).

````python
# Évaluer le meilleur modèle
gs_model.score(X_test, y_test)
````

Magnifique !

En utilisant `GridSearchCV`, nous constatons une belle amélioration du score de notre modèle.

Et le meilleur, c'est que, comme tout est dans un `Pipeline`, nous pourrions facilement reproduire ces résultats.
